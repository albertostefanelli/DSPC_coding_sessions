## Why these coding sessions matter in the age of AI

AI can produce code, tables, charts, and explanations. To use them well, we need to understand the code, read the data, interpret the visualizations, and decide what deserves to be presented. **These coding sessions are designed to build those skills.**

| What we need to do | What we will practice |

| **Decide what to present and how** | Select the comparison that matters and choose a clear number, table, or chart. Check axes, scales, units, and uncertainty before presenting a visualization. |

## 3. Give your inputs names

An assignment evaluates the expression on the right of `=` and gives its value the name on the left. We can use that name in later calculations. Assignment alone does not display the value; use the name as the last expression or call `print`.

Use descriptive names with letters, digits, and underscores. Names cannot start with a digit or contain spaces, and capitalization matters: `budget` and `Budget` are different names. A `#` starts a comment for human readers.

```python
budget = 24000
fixed_cost = 6000
marginal_cost_per_impression = 0.10
budget
```

### Fixed, marginal, and average cost

- **Fixed cost:** The setup expense, paid before any impressions are delivered.
- **Marginal cost:** The additional cost of delivering one more impression.
- **Average cost:** Total spending divided by impressions delivered, including setup.

We assume the full budget is spent, the budget exceeds the fixed cost, and marginal cost stays positive and constant. These practice values give whole numbers of impressions.

```python
spend_available_for_delivery = budget - fixed_cost
impressions = spend_available_for_delivery / marginal_cost_per_impression
average_cost_per_impression = budget / impressions

print("Impressions delivered:", impressions)
print("Average dollars per impression:", average_cost_per_impression)
```

### Calling built-in functions

A function call uses a name followed by parentheses and arguments. We have used `print` to display results. `round(number, decimal_places)` returns a rounded number; it is useful for readable output.

Keep the unrounded value for calculations and round when displaying it.

```python
print("Marginal dollars per impression:", marginal_cost_per_impression)
print("Average dollars per impression:", round(average_cost_per_impression, 4))
```

**Discuss:** Why is average cost about 0.1333 when the marginal cost is 0.10? Where does the difference come from?

## 4. Change an assumption

**Predict:** If the budget rises to USD 36,000 but both costs stay the same, will the number of impressions rise or fall? What will happen to average cost?

Run the next cell. Notice that assigning a new budget does **not** recalculate the previously assigned value of `impressions`. Variables hold calculated values, not live spreadsheet formulas.

```python
budget = 36000
print("New budget:", budget)
print("Previously calculated impressions:", impressions)
```

Now rerun the dependent calculations. You can rerun their earlier cell or run the version below. The result should reflect the new budget.

```python
spend_available_for_delivery = budget - fixed_cost
impressions = spend_available_for_delivery / marginal_cost_per_impression
average_cost_per_impression = budget / impressions

print("Updated impressions:", impressions)
print("Updated average dollars per impression:", round(average_cost_per_impression, 4))
```

**Check:** You should get 300,000 impressions at an average cost of 0.12 each. The fixed cost is spread across more impressions. Marginal cost has not changed.

## Optional practice: absolute differences and nested calls

If you have extra time, `abs(x)` returns an absolute value, `min(...)` returns the smallest argument, and `max(...)` returns the largest. A returned value can be assigned to a name or passed to another function.

Suppose three teams complete 120, 95, and 140 conversations. The next cell finds the smallest difference between any pair. Read nested calls from the inside out: calculate each subtraction, take its absolute value, then select the minimum.

```python
team_a_conversations = 120
team_b_conversations = 95
team_c_conversations = 140

smallest_gap = min(
    abs(team_a_conversations - team_b_conversations),
    abs(team_a_conversations - team_c_conversations),
    abs(team_b_conversations - team_c_conversations),
)
print("Smallest gap in completed conversations:", smallest_gap)
```
