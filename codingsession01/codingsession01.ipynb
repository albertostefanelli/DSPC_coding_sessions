## 2. Arithmetic operators

**Arithmetic Operators in Python**
- `+` addition  
- `-` subtraction  
- `*` multiplication  
- `/` division  
- `**` exponentiation  

Suppose a company reports its quarterly revenues (in millions of dollars) as follows:

- Q1: 120
- Q2: 135
- Q3: 150
- Q4: 130

To calculate the company’s total annual revenue, we add up the revenue from all four quarters:

```python
# total annual revenue (in millions)
120 + 135 + 150 + 130
```

**Order of Operations**  
Python follows the standard order of operations (PEMDAS/BODMAS). Use parentheses `()` to control evaluation.

Suppose that in addition to quarterly revenues we also have data on the annual costs incurred by the company
- Operating costs = 80
- Administrative costs = 95

To calculate the annual profit, we first compute the total annual revenue and then subtract the total annual costs:

```python
# annual profit (in millions)
(120 + 135 + 150 + 130) - (80 + 95)
```

**Tip:** When in doubt, add parentheses. It makes the order of operations clear and the code easier to read.

## 4. Embedded Python functions

In Python, functions are like small machines: they take some input, perform a task, and give you something back. Python already comes with many built-in (embedded) functions that you can use right away. To use a function:

- Write the function name (e.g., print, len, sum).
- Add parentheses () after the name.
- Provide the input. If you don’t provide it, Python will usaully give you an error.

```python
# Calling built-in functions
abs(-12)            # 12
min(4, 9, 2)        # 2
max(3, 7, 5)        # 7
```

**Additional opions:** Some functions require at least one argument, but they also allow you to specify additional options. If you don’t provide those extra arguments, Python uses a default value.





```python
# round() requires a number (the value to round)
print(round(5.678))      # → 6 (default: rounds to nearest whole number)

# round() also accepts an optional second argument:
# the number of decimal places to keep
print(round(5.678, 2))   # → 5.68 (rounds to 2 decimal places)
print(round(5.678, 1))   # → 5.7  (rounds to 1 decimal place)
```

Another example is the `sorted()` function which can be used to arrange items in order. It creates a new sorted list from the items you give it, without changing the original.


```python
numbers = [5, 2, 9, 1]

# Default: sorts in ascending order
print(sorted(numbers))          # → [1, 2, 5, 9]

# Optional argument: reverse=True
print(sorted(numbers, reverse=True))  # → [9, 5, 2, 1]
```

**Tip:** Whenever you encounter a new function, try running `help(function_name)`. This lets you look inside functions and objects. It shows:
- What the function does
- What arguments it accepts
- Sometimes, examples of how to use it

```python
print("sorted help()")
help(sorted)
print("abs help()")
help(abs)
```

**Nested functions**: When functions are nested (one inside another), Python evaluates them inside-out. That means the inner function runs first, and its result is passed to the outer function.



```python
your_score = 80.2
your_enemy = 78.5
your_friend = 85.3

# Distance to nearest neighbor score
nearest_distance = min(abs(your_score - your_enemy), abs(your_score - your_friend))
nearest_distance
```

**Tip:** When you see a function inside another function, read it like layers:

- Start with the innermost function
- Pass the result to the outer function.
- Keep moving outward until the full expression is done.

## Error safari

Python is very strict about how code must be written.
To practice, we’ll use these blocks as a live **error safari**:

* The **broken code** is commented out so the notebook still runs.
* The **uncommented code** shows the corrected version.

**1. Forgetting parentheses in function calls**

```python
# Bad
# print "Hello"

# Good:
print("Hello")
```

**2. Using a variable before assignment (NameError)**


```python
# Bad (NameError):
# print(result)

# Good:
result = 10 + 5
print(result)
```

**3. Case sensitivity**

```python
# Bad (NameError):
# Name = "Alice"
# print(name)

# Good:
name = "Alice"
print(name)
```

**4. Overwriting built-in functions**

```python
# Bad (shadows built-in function):
# sum = 10 + 5
# print(sum([1,2,3]))  # would error

# Good:
total = 10 + 5
print(sum([1, 2, 3]))
```

**5. Integer division vs. float division**

```python
# In Python / is float division, // is integer division
five_over_two_float = 5 / 2   # 2.5
five_over_two_int   = 5 // 2  # 2
five_over_two_float, five_over_two_int
```
