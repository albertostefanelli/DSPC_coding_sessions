# 1. From Curiosity to Research Design

This guide helps you move from an initial curiosity to a structured, theory-driven research design.  

## 1.1) Real-World Curiosity  
Identify something important, unresolved, or puzzling in politics that *genuinely* interests you. A good research question starts with curiosity and relevance, not just data availability.

*Example:*  
Why do some political candidates attract a large amount of individual donations, while others struggle to raise funds — even within the same party?

---

## 1.2) Narrow to Something Concrete  
Refine your curiosity by specifying focussing on something concreate and specific, along with one measurable **outcome**. Be specific enough to design a testable study but broad enough to matter.

*Example:*  

Do more ideologically extreme candidates raise fewer dollars per donor during general-election campaigns?

---

## 1.3) Identify the Puzzle  
Explain **why the answer isn’t obvious**.  What competing forces might lead to different outcomes? Where does prior evidence disagree?

*Example:*  

- Extremes attract media attention and loyal activists (which could help fundraising), but many voters are moderates and organizations value access and predictability.  
- Since 2016, however, some extreme candidates (e.g., Donald Trump, Marjorie Taylor Greene) have raised massive sums from small donors.

---

## 1.4) Check Relevance (hit 2 of 3)  

Ask whether your question matters beyond your own curiosity or personal interest

- **Practical:** Would campaigns or policymakers care?  
  *Example:* Campaigns need to predict fundraising potential before deciding to run or target districts.  
- **Scholarly:** Does it engage an ongoing debate or fill a gap?  
  *Example:* It speaks to debates about polarization, money in politics, and representation.  
- **Civic:** Does it matter for democracy or equality?  
  *Example:* If only extremists can raise viable funds, it affects who represents voters in Congress.

*Rule of thumb:* Aim to hit at least **two** of these dimensions.

---

## 1.5) Do a Quick Literature Scan  
Use tools like **Google Scholar**, **JSTOR**, or **SSRN** to see what’s already known.

Take notes on:
- **Measures** used (e.g., CFscores for ideology, FEC receipts for donations).  
- **Designs** (observational, experimental, natural experiments).  
- **Consensus vs. disagreement**, and where clear **gaps** exist (e.g., time periods, offices, donor types).  

*Example based on literature:*  
Scholars like **Ansolabehere et al. (2003)** and **La Raja & Schaffner (2015)** note that donors prefer moderate candidates.  
In contrast, **Barber, Canes-Wrone & Thrower (2017)** and **Bonica (2017)** find that online small donors favor ideological extremes.  

---

## 1.6) Define Scope & Scope Conditions  

**A. Universe & Unit**  
- **Universe:** Define geography, office, time, and stage (and justify why).  
  *Example:* U.S. House general elections, 1996–2024.
- **Unit of analysis:** The smallest entity that matches your theory.  
  *Example:* Candidate–election (each candidate’s donation in each election year).

**B. Scope Conditions (Theory Limits)**  
State **when** your mechanism should hold and **when** it should not.  

*Example:* The relationship between ideology and fundraising should be strongest when PACs dominate (pre-2016) and weaker after 2016, when online small-donor platforms gained power.

**Scope Statement:**  

> We study candidates for the U.S. House of Representatives over the past 30 years. We exclude Senators because statewide contests differ substantially in electorate size, media-market reach, six-year electoral cycles, and donor composition, all of which would confound comparability with single-district House races. Our analysis focuses on major-party nominees—Democrats and Republicans—at the candidate–election level. Third-party and independent candidates are excluded due to their limited presence and distinct fundraising dynamics. We also exclude unopposed races, candidates who self-fund more than 70% of total receipts, and special elections, where atypical timing and donor mobilization could bias results.

---

## 1.7) Set Expectations and Hypotheses  
Turn your theory into **testable claims** grounded in logic and **existing literature**.  Your hypotheses should explain *why* you expect something to happen — not just *what*.

This step is based on **theoretical reasoning** and **prior research**, not personal opinion.   

- **Ansolabehere, de Figueiredo & Snyder (2003)** — "Why is There So Little Money in U.S. Politics?"  
- **Bonica (2013, 2017)** — CFscores and ideological donor behavior.  
- **Barber, Canes-Wrone & Thrower (2017)** — Donor ideology and candidate positioning.  
- **La Raja & Schaffner (2015)** — How campaign finance laws shape polarization.

*Example hypotheses:*  

- *H1:* More ideologically extreme candidates attract fewer dollars per donor, as moderate and viable candidates appeal more broadly to individual donors.
- *H2a* (Pre-2016): The negative relationship between ideological extremity and dollars per donor is strongest before 2016 due to the appeal of moderate candidates.
- *H2b* (Post-2016): After 2016, the negative association weakens or reverses, as small-donor platforms and ideological mobilization make extreme candidates more financially competitive.


**Tip:** You may have multiple reasons why one of your hypotheses could be true.  Discuss each in your theoretical section, and explain **why one mechanism may be more important than another**, or why both might be equally plausible. If you cannot rule out one explanation in favor of the other, acknowledge this as a **limitation** (see below).

---

## 1.8) Define Outcomes & Predictors (Measurable!)  

Your theory should translate into clearly measurable outcomes (dependent variables) and predictors (independent variables) that capture the key concepts you want to test. Outcomes reflect what you’re trying to explain or predict. Predictors represent the theoretical drivers behind those outcomes.

Define each variable in observable, quantifiable terms, and specify how it will be measured or coded (e.g., continuous, categorical, or binary).

*Example:*  
- **Outcome:** Total individaul contributions raised in the general-election period divided by the numer of unique donors.
- **Key predictor:** A categorical measure of ideological extremity (e.g., distance from the party median or on the overall distribution), a dummy variable that define whether it is pre or post 2016.
- **Controls:** Incumbency, gender, party ...

**Tip:** It can be difficult to find the exact variable you need, so you may use a *proximate variable* as a substitute but be explicit about your theoretical expectations.  

Remember: **total dollars received** is not the same as **dollars per donor**; each measures a different aspect of fundraising and answers a different research question.

---

## 1.9) Data Plan

When you design a study, you need to explain where your data come from and how you will combine them so that every variable in your theory can be measured consistently. This often means pulling data from several public sources and mergining them together in a way that makese sense.

*Example Sources:*  

- *FEC filings* — candidate receipts and donor types.  
- *OpenSecrets* — classify donors (individual vs. PAC).  
- *Voteview/CFscores* — ideology measures.  
- *MIT Election Lab* — election results and competitiveness.  
- *Bonica Dataset* — ideology and money in politics data.

**Adjustments:**  When researchers work with real-world data, not every variable is available or consistent across all years or places. Sometimes data are missing, measured differently, or only start being collected at a certain time.

Instead of giving up on the research question, we make adjustments — careful, transparent decisions that keep the analysis valid while acknowledging the limitations of the data.

These adjustments are not about "fixing" the data. They are about making the best use of the evidence that exists in a way that still tests the theory fairly.

*Example:* If district-level small-donor data are incomplete before 2002, restrict the sample (2002–2024) or substitute total individual receipts as a proxy.

---

## 1.10) Modeling  

**Goal:** Test whether your predictor (e.g., *Extremity*) is linked to your outcome (e.g., *Dollars per donor*).  

- **Model type:** Match to data type
  - continuous outcome: OLS/mean comparison
  - binary outcome: logistic regression/cross tables/proprtions

*Example:*  

- Compare mean **dollars per donor** between **extreme** and **moderate** candidates; test the difference with a **permutation test** (shuffling labels).  
- Then estimate an **OLS regression** controlling for incumbency, seat type, gender, and party.   

## 1.11) Acknowledge Limitations

Every research design has limitation that can be [...] acknowledging them strengthens your credibility and helps others interpret your findings appropriately.

**Common types of limitations:**

- **Measurement limits:** Some variables (e.g., ideology, “extremity,” or donor motivation) are imperfectly captured by available proxies. You can also use robustness checks to alievates some of these limitations
  - *Example:* If ideology is measured from donation patterns, and you then use it to explain donations, you’re essentially using donations to predict donations. Maybe you can use the DW-NOMINATE measure of ideology to check whether your results hold.
- **Causal inference:** Observational data may show correlations that do not prove causation.  
  - *Example:* Extreme candidates might raise fewer dollars not because of ideology itself, but because they run in less competitive districts.  
- **Data availability:** Missing or incomplete data may make your finding less generalisable.
  - *Example:* lack of consistent measure of ideology for challangers
- **External validity:** Findings from U.S. House races may not generalize to other contexts
  - *Example:* Not accurate insights on senate, local, or international elections.

**Tip:** Be explicit about what your design can and cannot claim.  A strong paper explains its trade-offs openly. For instance, "these results describe associations, not causal effects" or noting where future research could test the mechanism more directly.

## 2.7) Construct your DV(s)  

In this project, we are interested in the **dollar amount per donor** received by each candidate.  
To construct this variable, we need three inputs:

- `total_unitemized` – total dollars from **unitemized** individual contributions (small donors below the reporting threshold).  
- `total_indiv_contribs` – total dollars from **itemized** individual contributions.  
- `num_givers` – number of **unique individual donors** giving to the candidate.

Conceptually:

$$
\text{dollars_per_donor} = \frac{\text{total_unitemized} + \text{total_indiv_contribs}}{\text{num_givers}}
$$

**Importantly:** we need to filter out any candidates with missing donation data — that is, rows where key fundraising variables contain NA values. In this example, we happen to have no missing data, but in your own project you might. In that case, you can remove incomplete observations using `dime_subset.dropna(subset=["total_indiv_contribs", "total_unitemized", "num_givers"])`

```python
# 1. Create a new variable called "individual_total"
#    This adds together itemized and unitemized individual contributions
#    to get the total amount raised from individuals.
dime_subset["individual_total"] = (
    dime_subset["total_indiv_contribs"] +
    dime_subset["total_unitemized"]
)

# 2. Compute "dollars per donor" (average amount given per donor)
#    Start by creating the column and filling it with missing values (pd.NA)
#    so we can assign real values later.
dime_subset["dollars_per_donor"] = pd.NA  # start with missing

# 3. Create a filter (mask) that keeps only rows where the number of givers is greater than 0
#    This prevents dividing by zero, which would cause an error.
valid_mask = dime_subset["num_givers"] > 0

# 4. For rows that pass the filter, calculate dollars_per_donor
#    by dividing total individual contributions by the number of givers.
dime_subset.loc[valid_mask, "dollars_per_donor"] = (
    dime_subset.loc[valid_mask, "individual_total"] /
    dime_subset.loc[valid_mask, "num_givers"]
)
```

### 2.9.1) Operationalizing Ideological Extremity


Now we need to operationalize ideological extremity — that is, turn the abstract idea of how “far” a candidate is from the ideological center into something we can actually measure with data.

There are several possible approaches:

- **Follow the literature:** The most rigorous approach is to first review how other scholars have defined and measured extremity. Many studies use CFscores (Bonica 2013) or DW-NOMINATE scores and calculate extremity as the absolute distance from the ideological center (e.g., abs(CFscore) or abs(DW-NOMINATE)). However, different papers define “center” differently — some use the overall distribution of candidates, while others measure extremity within each party
- **Create categories based on the statistical propety of the distribution:** You can classify candidates into groups such as “moderate,” “somewhat extreme,” and “very extreme” using quartiles, percentiles, or z-scores of their ideology scores. This is especially useful for descriptive or visualization purposes.
- **Your own categorization:** You can also develop your own measure or classification of extremity, tailored to your specific research question. For instance, you might decide on custom thresholds or combine multiple indicators (like ideology scores and issue positions). This approach requires extensive justification. You need to clearly explain why your method makes theoretical and empirical sense.

```python
def classify_ideology(score):
    if pd.isna(score):
        return None
    elif -0.5 <= score <= 0.5:
        return "Centrist"
    elif (-0.8 <= score < -0.5) or (0.5 < score <= 0.8):
        return "Soft Partisan"
    elif (-1.5 <= score < -0.8) or (0.8 < score <= 1.5):
        return "Mainstream"
    else:
        return "Extreme"

dime_subset["ideology_category"] = dime_subset["dwdime"].apply(classify_ideology)
```

### 2.10.2) H1 - Mean Comparison


```python
# Function to calculate the mean difference amd permutation tests seen in prevous labs
def mean_diff(data, group_var, outcome_var):
    """
    Calculates the difference in mean outcome between two groups.

    Parameters:
    - data: pandas DataFrame
    - group_var: column name indicating the grouping variable (e.g., 0/1 or category)
    - outcome_var: column name for the outcome variable

    Returns:
    - Mean difference (Group 1 - Group 0)
    """
    return data.groupby(group_var)[outcome_var].mean().diff().iloc[-1]

    # Function to perform a permutation test for mean differences
def permutation_test(data, group_var, outcome_var, n_permutations=1000):
    observed_diff = mean_diff(data, group_var, outcome_var)
    permuted_diffs = []

    # Run the permutation test n_permutations times
    for _ in range(n_permutations):
        # Shuffle the grouping variable
        shuffled_group = np.random.permutation(data[group_var])

        # Create a copy of the data with shuffled group assignments
        permuted_data = data.copy()
        permuted_data[group_var] = shuffled_group

        # Calculate the mean difference for each permutation
        permuted_diff = mean_diff(permuted_data, group_var, outcome_var)
        permuted_diffs.append(permuted_diff)

    # Calculate the p-value as the proportion of permuted differences
    # as "extreme" as the observed mean difference
    p_value = np.mean(np.abs(permuted_diffs) >= np.abs(observed_diff))

    return p_value

```

```python
# Prepare results container
results = []

# Define baseline group (centrists)
baseline = "Centrist"

# Define comparison groups
groups = ["Soft Partisan", "Mainstream", "Extreme"]

for g in groups:
    # Subset data for the two groups being compared
    subset = dime_subset[dime_subset["ideology_category"].isin([baseline, g])].copy()

    # Compute observed mean difference
    observed_diff = mean_diff(subset, "ideology_category", "log_dollars_per_donor")

    # Compute p-value via permutation test
    p_val = permutation_test(subset, "ideology_category", "log_dollars_per_donor", n_permutations=1000)

    # Store results
    results.append({
        "Comparison": f"{baseline} vs {g}",
        "Mean (Centrist)": subset.loc[subset["ideology_category"] == baseline, "log_dollars_per_donor"].mean(),
        "Mean (Other)": subset.loc[subset["ideology_category"] == g, "log_dollars_per_donor"].mean(),
        "N (Centrist)": subset.loc[subset["ideology_category"] == baseline, "log_dollars_per_donor"].count(),
        "N (Other)": subset.loc[subset["ideology_category"] == g, "log_dollars_per_donor"].count(),
        "Mean Difference (log)": observed_diff,
        "Mean Difference (%)": np.exp(observed_diff) - 1,
        "Permutation p-value": p_val
    })

# Convert results to DataFrame
perm_results = pd.DataFrame(results)
print(perm_results)
```

Because our dependent variable is in natural logs, these mean differences represent proportional (percentage) changes rather than absolute dollar gaps.
To interpret them in percentage terms, we exponentiate the difference:
$\text{% change} = (e^{\text{difference}} - 1) \times 100$

The difference between centrists and mainstream candidates is both large and statistically significant (p < 0.001). On average, mainstream candidates raise roughly 30% less per donor than centrists.

Soft partisans and extremes:
The smaller gaps (–6% and –9%) are not statistically significant, meaning the fundraising capacity of soft partisans, and extreme candidates is roughly comparable to centrists

### 2.10.2) H2 - Mean Comparison

Now we move to our over time hypthesis about. We are testing whether the nagative mean difference between centrists and extrests has become less negative after 2016


```python
# Prepare results container
results = []

# Define baseline and comparison groups
baseline = "Centrist"
groups = ["Soft Partisan", "Mainstream", "Extreme"]

# Loop through each period (pre vs post 2016)
for period, label in [(0, "Pre-2016"), (1, "Post-2016")]:
    # Subset data for the given period
    data_period = dime_subset[dime_subset["post_2016"] == period].copy()

    for g in groups:
        # Subset for only the two ideological groups being compared
        subset = data_period[data_period["ideology_category"].isin([baseline, g])].copy()

        # Compute observed mean difference
        observed_diff = mean_diff(subset, "ideology_category", "log_dollars_per_donor")

        # Compute permutation-based p-value
        p_val = permutation_test(
            subset, "ideology_category", "log_dollars_per_donor",
            n_permutations=1000
        )

        # Store results
        results.append({
            "Period": label,
            "Comparison": f"{baseline} vs {g}",
            "Mean (Centrist)": subset.loc[
                subset["ideology_category"] == baseline, "log_dollars_per_donor"
            ].mean(),
            "Mean (Other)": subset.loc[
                subset["ideology_category"] == g, "log_dollars_per_donor"
            ].mean(),
            "N (Centrist)": subset.loc[
                subset["ideology_category"] == baseline, "log_dollars_per_donor"
            ].count(),
            "N (Other)": subset.loc[
                subset["ideology_category"] == g, "log_dollars_per_donor"
            ].count(),
            "Mean Difference (log)": observed_diff,
            "Mean Difference (%)": np.exp(observed_diff) - 1,
            "Permutation p-value": p_val
        })

# Convert to DataFrame
perm_results = pd.DataFrame(results)
print(perm_results)
```

This suggests that extreme candidates appear far more capable of mobilizing donors after 2016. ### 2.10.3) H1 - Regression

As mentioned earlier, we also run a regression analysis to check whether our mean comparison results are robust to the inclusion of control variables that could bias the observed relationship between our dependent and independent variables.

Adding control variables allows us to account for other factors that might  affect fundraising.

It is **essential to justify** each control you include:

- Incumbency: Incumbent candidates generally raise more money because they benefit from established donor networks and greater name recognition. If incumbents also tend to be more moderate, then the apparent fundraising advantage of moderates (or disadvantage of extremists) may simply reflect incumbency rather than ideology.
- Party: If one party experiences a sharper rise in extremist candidates or relies more heavily on small-dollar donors, this could confound the link between extremism and total fundraising.
- Gender: Male and female candidates may face different donor expectations that interact with their extremity. Apparent ideological effects could in part be driven by gender differences rather than extremism itself.
- Time trend: If the number of ideologically extreme candidates has risen substantially after 2016 and the average dollar amount per donor has simultaneously declined---perhaps due to the growth of small-dollar online fundraising platforms---then the apparent relationship between extremism and donations may partly reflect these time-based changes rather than ideological effects per se.

Before estimating the regression, we need to “dummy-code” (or one-hot encode) our categorical independent variables, such as the ideology categories.

Regression models like OLS can only handle numeric variables, not text or labels. By converting each category into a binary variable (0 or 1), we can estimate how each ideological group differs from a reference category (for example, “Centrist”).


```python
## PARTY (100 = Democrat, 200 = Republican)
dime_subset['party_dem'] = (dime_subset['party'] == 100).astype(int)
dime_subset['party_rep'] = (dime_subset['party'] == 200).astype(int)

## GENDER (F/M)
dime_subset['female'] = (dime_subset['cand_gender'] == 'F').astype(int)
dime_subset['male'] = (dime_subset['cand_gender'] == 'M').astype(int)

## INCUMBENCY STATUS (I = Incumbent, C = Challenger, O = Open)
dime_subset['incumbent'] = (dime_subset['ico_status'] == 'I').astype(int)
dime_subset['challenger'] = (dime_subset['ico_status'] == 'C').astype(int)
dime_subset['open'] = (dime_subset['ico_status'] == 'O').astype(int)

# CATEGORICAL MEASURE OF IDEOLOGY
dime_subset['ideo_centrist'] = (dime_subset['ideology_category'] == 'Centrist').astype(int)
dime_subset['ideo_soft'] = (dime_subset['ideology_category'] == 'Soft Partisan').astype(int)
dime_subset['ideo_main'] = (dime_subset['ideology_category'] == 'Mainstream').astype(int)
dime_subset['ideo_extreme'] = (dime_subset['ideology_category'] == 'Extreme').astype(int)
```

Before fitting our regression, we must drop one category (called the reference category) to avoid perfect multicollinearity — a situation where one variable can be exactly predicted from the others.

The reference category becomes the baseline against which all other groups are compared.

For instance, if “Centrist” is the reference, the coefficients for “Soft Partisan,” “Mainstream,” and “Extreme” will show how much higher or lower their average donations are relative to centrists.

```python
# 1) Select only the columns we need
model_df = dime_subset[[
    'log_dollars_per_donor',
    'ideo_soft', 'ideo_main', 'ideo_extreme',
    'party_dem',
    'female',
    'incumbent', 'challenger',
    'cycle'
]].copy()

# 2) Make sure everything is numeric
model_df = model_df.apply(pd.to_numeric, errors='coerce')

# 3) Drop rows with any missing values in these columns
model_df = model_df.dropna()

# 4) Define X and y
y = model_df['log_dollars_per_donor']
X = model_df[[
    'ideo_soft', 'ideo_main', 'ideo_extreme',
    'party_dem',
    'female',
    'incumbent', 'challenger',
    'cycle'
]]

# 5) Add constant and run the model
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

As seen before, the coefficient for extreme candidates (+0.06) is positive but not statistically significant (p = 0.15), meaning that — after controlling for other factors — extreme candidates raise about 6% more per donor than centrists, but the evidence isn’t strong enough

### 2.10.4) H2 - Regression

```python
# Prepare an empty dictionary to store results
models = {}

# Loop over both time periods: pre-2016 (0) and post-2016 (1)
for period, label in [(0, "Pre-2016"), (1, "Post-2016")]:
    print(f"\n=== Regression for {label} elections ===")

    # 1) Subset the data by period
    subset = dime_subset[dime_subset["post_2016"] == period].copy()

    # 2) Select only the columns needed for the model
    model_df = subset[[
        'log_dollars_per_donor',
        'ideo_soft', 'ideo_main', 'ideo_extreme',
        'party_dem',
        'female',
        'incumbent', 'challenger',
        'cycle'
    ]].copy()

    # 3) Ensure all variables are numeric
    model_df = model_df.apply(pd.to_numeric, errors='coerce')

    # 4) Drop rows with missing data
    model_df = model_df.dropna()

    # 5) Define dependent and independent variables
    y = model_df['log_dollars_per_donor']
    X = model_df[[
        'ideo_soft', 'ideo_main', 'ideo_extreme',
        'party_dem',
        'female',
        'incumbent', 'challenger',
        'cycle'
    ]]

    # 6) Add constant term and fit OLS model
    X = sm.add_constant(X)
    model = sm.OLS(y, X).fit()

    # 7) Save and display results
    models[label] = model
    print(model.summary())
```

Before 2016, there is no significant difference between centrists and extreme candidates in dollars raised per donor. After 2016, the pattern flips direction: the coefficient for extreme candidates becomes negative (–0.11), suggesting they raise roughly 11% less per donor than centrists on average. However, the relationship is still not significant at .05.

**So why did the permutation test earlier show a strong positive difference, while the regression does not?**

This happens because the regression includes control variables that adjust for factors likely to confound the relationship between ideology and fundraising — things like time trends, party affiliation, and incumbent status. Once these controls are added, the apparent fundraising advantage of extreme candidates disappears, suggesting that it was spurious — driven by correlated factors rather than ideology itself.
