## 1. Why do polls of the same race disagree?

### The story

Eight weeks before Election Day, three polls of the same state Senate race are released:

| Poll | How people were contacted | Who was surveyed | Sample size | Result |
|---|---|---|---|---|
| Landmark Research | Live phone calls | Likely voters | about 1,000 | Rep +4 |
| Progressive Data Institute | Online panel | Likely voters | about 1,000 | Dem +8 |
| State University Poll | Phone and online, then weighted | Registered voters | about 1,200 | Dem +1 |

An **online panel** is a group of people who signed up in advance to take surveys on the internet. Every number in the last column is a **lead**: one candidate's share of respondents minus the other's. A poll's headline numbers (each candidate's share and the lead) are called its **toplines**. The phone and online polls are 12 points apart.

**Units.** Shares are percentages (46.2%). Leads and differences between shares are **percentage points** (pp, or just "points"). Python gives you proportions (0.462); multiply by 100 to report either. We write a Democratic lead as a positive number and a Republican lead as a negative one, so "Rep +4" is −4. Before deciding which one to trust, we need to know how far apart two honest polls of the same race could land just by chance, and what else could be going on.

### Five reasons two polls disagree

Three reasons come from **collecting** the data:

- **Sampling error.** A poll interviews about 1,000 people to learn what hundreds of thousands think. Interview a different 1,000 and you get a different number. This is random noise.
- **Nonresponse bias.** The people who answer the survey differ systematically from the people who don't, in ways that affect the answer. If Republican-leaning voters are less likely to pick up the phone, the poll overstates Democratic support, even if the pollster dialed exactly the right mix of numbers. Phone response rates are now below 5%, so the people who do answer are not a random sample. Phone polls tend to reach older voters, who answer calls from unknown numbers. Online panels tend to reach younger, more politically engaged people.
- **Mode effects.** The **mode** is how people are contacted: a live phone call, an automated call, an online panel, a text message with a link. A **mode effect** means the same person gives a different answer depending on the mode, for example saying what sounds good when a live interviewer is listening.

Two more come from **analyzing** the data:

- **Weighting.** Which groups the pollster adjusts for, and to what targets.
- **Likely-voter screen.** The pollster's rule for deciding which respondents count as voters: asking "Do you plan to vote?", checking past turnout in the voter file, or a mix of both. Different rules produce different electorates from the same interviews.

Only the first reason is random. The other four can push a poll in one direction.

### Two kinds of error: noise and bias

**Sampling error** is noise. It pushes a poll up or down at random, and it shrinks as the sample grows. **Bias** pushes every poll from the same method in the same direction, and it does **not** shrink when you collect more of the same kind of data. A poll of 50,000 biased respondents is a more precise wrong answer.

![Four panels of 200 simulated polls each. Top row, an unbiased pollster: polls cluster around the true Democratic share of 48 percent, widely with 500 respondents and tightly with 5,000. Bottom row, a pollster whose respondents lean Democratic: polls cluster around 51 percent, and with 5,000 respondents they cluster tightly around the wrong number.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig1_noise_vs_bias.png)

*Figure 1. Each dot is one simulated poll. A bigger sample tightens the cloud (left to right). It does not move the cloud back to the truth (bottom row).*

The **margin of error** printed next to a poll measures only the first kind of error. You can compute it from the poll itself. You cannot compute the bias from the poll; you can only argue about which direction it goes. In 2020, national presidential polls overstated Biden's margin by 3.9 points, and nearly all of them missed in the same direction. Averaging more of those polls would not have helped.

Bias can also come from the pollster. A **house effect** is a pollster's methods (how it contacts people, words questions, weights and screens for likely voters) consistently favoring one side. Polls paid for by a party or campaign deserve extra care for a second reason: the campaign decides which of its internal polls to release, and it tends to release the ones it likes. Picking the poll that tells you what you want to hear is **cherry-picking**, and it is how a set of honest polls ends up telling a dishonest story. Before trusting a poll, ask who paid for it, who was interviewed, how they were contacted, and how the results were weighted.

### Measuring sampling error: the margin of error

The margin of error is the "plus or minus" of a 95% confidence interval, the same idea as in Week 5, applied to a poll number instead of a treatment effect:

**margin of error = 1.96 × standard error**

**standard error = (how far a typical respondent is from the average) ÷ √n**

The bottom of the fraction is the sample size: four times the sample gives half the standard error. The top is how spread out people are on the question. For a yes/no question such as "Did you pick the Democrat?", the top is √(p × (1 − p)), where p is the share who say yes. In the phone poll, 46% of 1,001 respondents picked the Democrat:

> √(0.46 × 0.54) ÷ √1,001 = 0.50 ÷ 31.6 ≈ 0.016, or 1.6 points. The margin of error is 1.96 × 1.6 ≈ ±3 points, so "Dem 46%" means "Dem somewhere between about 43% and 49%".

**A lead is noisier than a share.** If one respondent switches from the Democrat to the Republican, the Democrat's share falls by one person and the Republican's rises by one, so the lead moves by two. To measure a lead, score each respondent **+1** for the Democrat, **−1** for the Republican and **0** for undecided. The average score is the lead. Almost everyone picks a side, so a typical score sits about 1 away from the average, against about 0.5 for a yes/no answer. Same sample size, twice the spread, twice the standard error:

| Question | Possible answers | Typical distance from the average | Standard error | Margin of error |
|---|---|---|---|---|
| "Did you pick the Democrat?" (a share) | 0 or 1 | about 0.5 | 1.6 points | ±3 points |
| "Which way do you lean?" (a lead) | −1, 0 or +1 | about 1.0 | 3.1 points | ±6 points |

![Left: the phone poll's Democratic share of 46 percent with a margin of error of plus or minus 3 points. Right: three leads with 95 percent intervals on an axis from Rep +12 to Dem +16. The phone poll, Rep +4, and the university poll, Dem +1, both cross the tie line and are statistical ties. The online poll, Dem +8, clears the tie line but its interval reaches down to about Dem +2.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig2_share_vs_lead.png)

*Figure 2. The same sample size gives a share a margin of about ±3 points and a lead a margin of about ±6. Two of the three leads are statistical ties.*

A lead whose 95% interval includes zero is a **statistical tie**: the poll cannot tell us who is ahead. This is the Week 5 significance test again: is the lead distinguishable from zero?

**Comparing two polls.** The difference between two independent polls has its own standard error: square each poll's standard error, add them, and take the square root. With 3.1 points each, √(3.1² + 3.1²) ≈ 4.4 points. So two polls of 1,000 people need to be about 1.96 × 4.4 ≈ 9 points apart before we can rule out chance as the whole explanation.

### Fixing what you can: post-stratification

**Post-stratification** means adjusting the sample after the interviews are done so that it matches the population on characteristics we know from the census or the voter file, such as age, race, education and party registration. If the sample has too many older people, each younger respondent counts for more.

1. For each group, compute a **weight = population share of the group ÷ sample share of the group**.
2. Recompute the results, counting each respondent as many times as their weight.

A **weighted mean** is the sum of (answer × weight) divided by the sum of the weights.

> Two respondents: one picked the Democrat and has weight 3, the other picked the Republican and has weight 1. The weighted Democratic share is (1 × 3 + 0 × 1) ÷ (3 + 1) = 3/4 = 75%. Unweighted, it would be 1/2 = 50%.

![Left: the phone poll's age mix next to the district's. The poll has 6 percent aged 18 to 29 against 18 percent in the district, and 41 percent aged 65 and over against 25 percent. The weights are 3.00, 1.47, 0.90 and 0.60. Right: 180 dots for the young voters a fair poll of 1,000 would include. 60 filled dots answered and now count three times each. 120 faded dots never answered and are still missing.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig3_what_weighting_fixes.png)

*Figure 3. Weighting swaps the poll's age mix for the district's (left). It cannot bring back the people who never answered (right).*

**What post-stratification cannot do.** It fixes imbalances on the variables you weight by. It does not fix imbalances *within* those groups. A fair poll of 1,000 in our district would include about 180 voters under 30. The phone poll reached 60. Weighting counts each of them three times; it does not find the 120 who are missing. If the 60 who picked up are the politically engaged young voters, weighting makes the engaged three times louder, and the disengaged are still missing.

**Why not weight on everything?** First, you need a population target, and there is no census of, say, "trusts institutions". Second, groups get small: split the 60 young respondents by education and then by four regions and you have about seven people per group. Third, small groups produce extreme weights, so one respondent can move the result. Which variables a pollster weights on matters a lot: good pollsters handed the same raw interviews can publish results several points apart.

**Weighting costs precision.** Unequal weights make an estimate noisier. The **effective sample size** is the size of an unweighted poll that would be just as precise:

**n_eff = (sum of the weights)² ÷ (sum of the squared weights)**

If every weight is 1, n_eff equals the number of respondents. The more unequal the weights, the smaller n_eff and the wider the margin of error.

### Two polls of the same race

We have the raw interviews behind two of the three polls in the story: the phone poll and the online poll. Remember that these are **teaching data**. The university poll's raw data are not available, so we only use its published result, Dem +1.

Each row is one **respondent**.

| Column | What it records | Values |
|---|---|---|
| `respondent_id` | Respondent number | 1, 2, 3, … |
| `age_group` | The respondent's age group | `18-29`, `30-44`, `45-64`, `65+` |
| `vote_intent` | Who the respondent plans to vote for | `dem`, `rep`, `undecided` |
| `poll` | Which poll the respondent is from | `phone_poll`, `online_poll` |

We also need the population we want the polls to describe. The district's voter file gives the age mix of all registered voters: 18.0% are 18–29, 25.0% are 30–44, 32.1% are 45–64 and 24.9% are 65 or older.

```python
# Import the tools we need: pandas for tables, numpy for numbers, matplotlib for plots.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read both polls straight from the course repository (you need to be online).
# Source: teaching data built for this course (see the data notes linked above).
# Python joins the two quoted strings inside the parentheses into one address.
phone = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/phone_poll.csv'
)
online = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/online_poll.csv'
)

# Age mix of ALL registered voters in the district, from the voter file.
# A pandas Series is a column with labels; here the labels are the age groups.
pop_age = pd.Series({'18-29': 0.180, '30-44': 0.250, '45-64': 0.321, '65+': 0.249})

print('Phone poll respondents: ', len(phone))
print('Online poll respondents:', len(online))
phone.head()
```

## Setup: rebuild the objects from Part 2

This cell repeats the demonstration in Part 2 without the printouts: the score columns, the age shares and weights from 2c–2d, and the poll archive with its error columns from 2e.

```python
# 2a: 0/1 columns and the +1/0/-1 score for both polls.
phone['is_dem'] = (phone['vote_intent'] == 'dem').astype(int)
phone['is_rep'] = (phone['vote_intent'] == 'rep').astype(int)
phone['score'] = phone['is_dem'] - phone['is_rep']
online['is_dem'] = (online['vote_intent'] == 'dem').astype(int)
online['is_rep'] = (online['vote_intent'] == 'rep').astype(int)
online['score'] = online['is_dem'] - online['is_rep']

lead_phone = phone['score'].mean()
lead_online = online['score'].mean()

# 2c-2d: age shares, registered-voter weights, weighted leads.
phone_age = phone['age_group'].value_counts(normalize=True)
online_age = online['age_group'].value_counts(normalize=True)
phone_weights = pop_age / phone_age
online_weights = pop_age / online_age
phone_lead_by_age = phone.groupby('age_group')['score'].mean()
online_lead_by_age = online.groupby('age_group')['score'].mean()
phone_weighted_lead = (phone_age * phone_weights * phone_lead_by_age).sum()
online_weighted_lead = (online_age * online_weights * online_lead_by_age).sum()

# 2e: the FiveThirtyEight poll archive, close races only, and its error columns.
polls = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/final_polls_1998_2022.csv'
)
polls = polls[polls['margin_actual'].abs() < 10].copy()
polls['error'] = polls['margin_poll'] - polls['margin_actual']
polls['abs_error'] = polls['error'].abs()
polls['moe_lead'] = 1.96 * 2 * np.sqrt(0.25 / polls['sample_size']) * 100
polls['outside_moe'] = polls['abs_error'] > polls['moe_lead']

print(f'Phone lead {100*lead_phone:+.1f} -> weighted {100*phone_weighted_lead:+.1f} pp')
print(f'Online lead {100*lead_online:+.1f} -> weighted {100*online_weighted_lead:+.1f} pp')
print(f'Close-race polls: {len(polls):,}; outside their own margin of error: {polls["outside_moe"].mean():.1%}')
```

### Exercise 1. Fifteen polls can't all be wrong?

In the last three weeks before the 2016 presidential election, the FiveThirtyEight archive has 15 polls of Wisconsin, and every one of them shows Clinton ahead. Imagine a Clinton staffer saying: *"Fifteen polls out of fifteen have us ahead, some by 12 points. Wisconsin is safe. Send our volunteers somewhere else."*

**Calculate:** keep the 15 Wisconsin polls in `wi` (supplied). Count how many are statistical ties, meaning the lead is smaller than the poll's own margin of error (`n_ties`). Then compute the average lead across the 15 polls (`avg_lead`) and compare it with the actual result, `margin_actual`.

**Interpret:** is "fifteen out of fifteen" strong evidence that Wisconsin is safe? What does the number of statistical ties tell you? Averaging 15 polls should cancel out much of their random noise: did the average land near the result? What does that tell you about the kind of error these polls shared?

```python
wi = polls[polls['race'] == '2016_Pres-G_WI']
print(wi[['pollster', 'poll_date', 'sample_size', 'margin_poll', 'moe_lead', 'margin_actual']].round(1))
print()

# A statistical tie: the lead is smaller than the poll's own margin of error.
n_ties = (wi['margin_poll'].abs() < wi['moe_lead']).sum()
print(f'Polls: {len(wi)}, statistical ties: {n_ties}')

# Average of the 15 leads against the actual result (Dem minus Rep, in points).
avg_lead = wi['margin_poll'].mean()
actual = wi['margin_actual'].iloc[0]
print(f'Average poll lead: {avg_lead:+.1f} pp; actual result: {actual:+.1f} pp; miss: {avg_lead - actual:+.1f} pp')
```

**Interpretation:** "Fifteen out of fifteen" sounds decisive, but **10 of the 15** polls were statistical ties by their own margin of error, so most of them could not tell who was ahead. The average of the 15 polls was **Clinton +5.4**, and Trump won by **0.8** points, a miss of about **6 points**. Averaging cancels noise that pushes polls in different directions, but here every poll missed the same way, so the error was shared bias (who answered, how the polls were weighted), not noise, and more polls just repeated it. Wisconsin should have been treated as a close race, not a safe one.

*Grading notes:* full credit needs (1) the count of ties and what a tie means, (2) the average against the result, and (3) the point that averaging removes random error but not error all polls share. A common mistake is to treat agreement among many polls as proof: polls can agree because they share the same bias.

### Exercise 2. Would a bigger poll have saved us?

A vendor pitches the campaign a poll of 3,000 likely voters: *"With 3,000 interviews our margin of error on the lead is under ±4 points. Our number will be within 4 points of the result."*

**Calculate:** split `polls` into large polls (`sample_size >= 1500`) and small polls (`sample_size < 600`). For each group, print the number of polls, the average `moe_lead`, the average `abs_error`, and the share of polls with `outside_moe` equal to True.

**Interpret:** what does a larger sample buy, and what does it not buy? Use the difference between noise and bias from Figure 1. Is the vendor's promise justified?

```python
large = polls[polls['sample_size'] >= 1500]
small = polls[polls['sample_size'] < 600]

print(f'Large (n >= 1,500): {len(large):,} polls, average margin of error ±{large["moe_lead"].mean():.1f} pp, '
      f'average miss {large["abs_error"].mean():.1f} pp, outside own margin {large["outside_moe"].mean():.1%}')
print(f'Small (n < 600):    {len(small):,} polls, average margin of error ±{small["moe_lead"].mean():.1f} pp, '
      f'average miss {small["abs_error"].mean():.1f} pp, outside own margin {small["outside_moe"].mean():.1%}')

# Robustness check for instructors: large polls are mostly presidential, small ones often House.
# Within each office, the average miss still barely depends on sample size.
print()
print(polls.groupby(['race_type', polls['sample_size'] >= 1500])['abs_error'].mean().round(1))
```

**Interpretation:** Large polls have a margin of error well under half that of small polls (about **±4** against **±9** points), but their average miss is only a little smaller (about **3.8** against **4.6** points). A bigger sample shrinks the noise, but it does nothing about bias from who responds, how they were contacted, and how the pollster weighted and screened. The vendor confuses the margin of error with the total error: about half of large polls missed by more than their ±4, so a poll of 3,000 can be a more precise wrong answer.

*Grading notes:* full credit needs (1) numbers showing the margin shrinks much faster than the miss, (2) the noise vs bias explanation, and (3) a verdict on the promise. Large polls are mostly presidential and small ones are often House races; the check at the bottom of the cell shows that within each office the miss still barely depends on sample size, so the conclusion holds. Students who say "large polls are no better" overstate it: the average miss does fall a little.

### Exercise 3. Which variable should you weight on?

The **Cooperative Election Study (CES)** is a large real survey of American adults, run online with people recruited from opt-in internet panels. The file below has the **50,117** respondents in its fall 2024 pre-election wave who said they would definitely vote or had already voted. For each, it records their age group, education, who they voted for in 2020, and who they planned to vote for in 2024 (`dem` = Harris, `rep` = Trump). Trump won the national popular vote by about **1.5 points** (49.8% to 48.3%), so the true lead is about −1.5.

**Source:** Cooperative Election Study, Common Content 2024 (Schaffner, Shih, Ansolabehere and Pope), Harvard Dataverse, [doi:10.7910/DVN/X11EP6](https://doi.org/10.7910/DVN/X11EP6), CC0. Our extract keeps likely voters and four questions; the [data notes](https://github.com/albertostefanelli/DSPC_coding_sessions/blob/master/wk08_trusting_polls/data/README.md) give the exact steps.

`ces_weight` is the survey's own full weight. The CES team built it to match Census figures on age, gender, education and race, then adjusted it further on voter registration, state, and **2020 and 2024 presidential vote**, among other things. The supplied targets below are the age and 2020-vote mix those weights imply for these likely voters. A pollster tells you: *"Our raw sample leans Democratic, but we weight by age like everyone else, so we're fine."*

**Calculate:** compute the unweighted lead (`ces_lead`). Then, as in 2d, weight by age with `age_targets` (`age_weighted_lead`) and, separately, by 2020 vote with `vote2020_targets` (`vote2020_weighted_lead`). Print all three next to the actual result. Optional: compute the lead with the CES's full weights, `ces_weight`, using the row-by-row formula from 2d.

**Interpret:** which weighting moves the estimate toward the result, and which does not? Explain why, using two things from the tables: how far the sample is from the target on each variable, and how differently the groups vote. What does this tell you about the pollster's claim, and about why polls miss?

```python
ces = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/ces_2024_likely_voters.csv'
)
ces['score'] = (ces['vote_intent'] == 'dem').astype(int) - (ces['vote_intent'] == 'rep').astype(int)
age_targets = pd.Series({'18-29': 0.130, '30-44': 0.229, '45-64': 0.345, '65+': 0.296})
vote2020_targets = pd.Series({'Biden': 0.460, 'Trump': 0.429, 'Did not vote': 0.094, 'Other': 0.017})
actual_lead = -0.015

ces_lead = ces['score'].mean()

# Weight by age: same three steps as 2d.
ces_age = ces['age_group'].value_counts(normalize=True)
age_weights = age_targets / ces_age
lead_by_age = ces.groupby('age_group')['score'].mean()
age_weighted_lead = (ces_age * age_weights * lead_by_age).sum()

# Weight by 2020 vote.
ces_vote2020 = ces['vote_2020'].value_counts(normalize=True)
vote2020_weights = vote2020_targets / ces_vote2020
lead_by_vote2020 = ces.groupby('vote_2020')['score'].mean()
vote2020_weighted_lead = (ces_vote2020 * vote2020_weights * lead_by_vote2020).sum()

# Optional: the CES's full weights, row by row.
full_weighted_lead = (ces['score'] * ces['ces_weight']).sum() / ces['ces_weight'].sum()

print(pd.DataFrame({'Sample': ces_age, 'Target': age_targets, 'Lead (pp)': 100 * lead_by_age}).round(3))
print()
print(pd.DataFrame({'Sample': ces_vote2020, 'Target': vote2020_targets, 'Lead (pp)': 100 * lead_by_vote2020}).round(3))
print()
print(f'Unweighted:              {100*ces_lead:+.1f} pp')
print(f'Weighted by age:         {100*age_weighted_lead:+.1f} pp')
print(f'Weighted by 2020 vote:   {100*vote2020_weighted_lead:+.1f} pp')
print(f'Full CES weights:        {100*full_weighted_lead:+.1f} pp')
print(f'Actual result:           {100*actual_lead:+.1f} pp')
```

**Interpretation:** Unweighted, the CES likely voters give **Harris +15.3**, against an actual result of **Trump +1.5**. Weighting by age makes it slightly *worse* (**Harris +16.8**), while weighting by 2020 vote moves it about **10 points** toward the result (**Harris +4.9**); the full CES weights get to about **Harris +0.6**. Weighting by age does fix a real imbalance (too few voters aged 30–44: 17% against 23%), but those younger voters lean toward Harris, so correcting it pushes the estimate further from the result: the sample's problem is not its age mix. 2020 vote helps a lot because the sample has too many Biden voters (51% against 46%) and too few Trump voters (36% against 43%), and those groups vote almost entirely in opposite directions. So the pollster's claim is wrong: matching on demographics does not fix a sample whose problem is *which kind of voter* answered, and the remaining gap shows that even good weights do not remove all the bias.

*Grading notes:* full credit needs (1) the four numbers (the optional fifth is a bonus), (2) the two-part reason a weighting variable matters (imbalance in the sample, and a difference in vote between its groups), and (3) the link to nonresponse: the people who answer online surveys differ politically, within demographic groups, from those who don't. Students should not conclude that age never matters; it matters when the sample is far off on age, as in the phone poll in 2d. Note the targets come from the CES's own weights, so the 2020-vote result is a best case, and the full weights are partly adjusted to the 2024 result itself, so their Harris +0.6 is a check on the weighting, not a forecast a pollster could have made.

Sponsored polls lean toward their sponsor. One reason is a **house effect**: the sponsor's pollster can make choices about wording, likely voters and weighting that favor its side. The other is **cherry-picking**: campaigns choose which internal polls to release, so the ones we see are the favorable ones.
