# Coding Session 08: Is the poll wrong, or just noisy?

## 1. Why do polls of the same race disagree?

Four reasons come from **collecting** the data:

- **Sampling error.** A poll interviews about 1,000 people to learn what hundreds of thousands think. Interview a different 1,000 and you get a different number. This is random noise.
- **Nonresponse bias.** The people who answer the survey differ systematically from the people who don't, in ways that affect the answer. 
- **Mode effects.** A **mode effect** means the same person gives a different answer depending on the mode, for example saying what sounds good when a live interviewer is listening.
- **Measurement error.** How a pollster asks questions to people, words questions. For instance, the pollster's rule for deciding which respondents count as voters.

One more come from **analyzing** the data:

- **Weighting.** Which groups the pollster adjusts for, and to what targets. And what which respondents count as voters.

Only the first reason is random. The other 4 can push a poll in one direction.

### Two kinds of error: noise and bias

- **Sampling error** is noise. It pushes a poll up or down at random, and it shrinks as the sample grows.
- **Bias** pushes every poll from the same method in the same direction, and it does **not** shrink when you collect more of the same kind of data. A poll of 50,000 biased respondents is a more precise wrong answer. 

![Four panels of 200 simulated polls each. Top row, an unbiased pollster: polls cluster around the true Democratic share of 48 percent, widely with 500 respondents and tightly with 5,000. Bottom row, a pollster whose respondents lean Democratic: polls cluster around 51 percent, and with 5,000 respondents they cluster tightly around the wrong number.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig1_noise_vs_bias.png)

*Figure 1. Each dot is one simulated poll. A bigger sample tightens the cloud (left to right). It does not move the cloud back to the truth (bottom row).*

The **margin of error** printed next to a poll measures only the first kind of error. You can compute it from the poll itself. 

### Measuring sampling error: the margin of error

The margin of error is the "plus or minus" of a 95% confidence interval, the same idea as in Week 5, applied to a poll number instead of a treatment effect:

**margin of error = 1.96 × standard error**

**standard error = (how far a typical respondent is from the average) ÷ √n**

The bottom of the fraction is the sample size: four times the sample gives half the standard error. The top is how spread out people are on the question. For a yes/no question such as "Did you pick the Democrat?", the top is √(p × (1 − p)), where p is the share who say yes. Take a phone poll in which 46% of 1,001 respondents picked the Democrat:

> √(0.46 × 0.54) ÷ √1,001 = 0.50 ÷ 31.6 ≈ 0.016, or 1.6 points. The margin of error is 1.96 × 1.6 ≈ ±3 points, so "Dem 46%" means "Dem somewhere between about 43% and 49%".

**A lead is noisier than a share.** If one respondent switches from the Democrat to the Republican, the Democrat's share falls by one person and the Republican's rises by one, so the lead moves by two. To measure a lead, score each respondent **+1** for the Democrat, **−1** for the Republican and **0** for undecided. The average score is the lead. Almost everyone picks a side, so a typical score sits about 1 away from the average, against about 0.5 for a yes/no answer. Same sample size, twice the spread, twice the standard error:

| Question | Possible answers | Typical distance from the average | Standard error | Margin of error |
|---|---|---|---|---|
| "Did you pick the Democrat?" (a share) | 0 or 1 | about 0.5 | 1.6 points | ±3 points |
| "Which way do you lean?" (a lead) | −1, 0 or +1 | about 1.0 | 3.1 points | ±6 points |

![Left: the phone poll's Democratic share of 46 percent with a margin of error of plus or minus 3 points. Right: three leads with 95 percent intervals on an axis from Rep +12 to Dem +16. The phone poll, Rep +4, and the university poll, Dem +1, both cross the tie line and are statistical ties. The online poll, Dem +8, clears the tie line but its interval reaches down to about Dem +2.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig2_share_vs_lead.png)

*Figure 2. The same sample size gives a share a margin of about ±3 points and a lead a margin of about ±6. The three leads are three polls of the same state Senate race, released within two weeks of each other; two of them are statistical ties.*

**Comparing two polls.** The difference between two independent polls has its own standard error: square each poll's standard error, add them, and take the square root. With 3.1 points each, √(3.1² + 3.1²) ≈ 4.4 points. So two polls of 1,000 people need to be about 1.96 × 4.4 ≈ 9 points apart before we can rule out chance as the whole explanation.

### Fixing what you can: post-stratification

**Post-stratification** means adjusting the sample after the interviews are done so that it matches the population on characteristics we know from the census or the voter file, such as age, race, education and party registration. If the sample has too many older people, each younger respondent counts for more.

1. For each group, compute a **weight = population share of the group ÷ sample share of the group**.
2. Recompute the results, counting each respondent as many times as their weight.

A **weighted mean** is the sum of (answer × weight) divided by the sum of the weights.

> Two respondents: one picked the Democrat and has weight 3, the other picked the Republican and has weight 1. The weighted Democratic share is (1 × 3 + 0 × 1) ÷ (3 + 1) = 3/4 = 75%. Unweighted, it would be 1/2 = 50%.

![Left: the phone poll's age mix next to the district's. The poll has 6 percent aged 18 to 29 against 18 percent in the district, and 41 percent aged 65 and over against 25 percent. The weights are 3.00, 1.47, 0.90 and 0.60. Right: 180 dots for the young voters a fair poll of 1,000 would include. 60 filled dots answered and now count three times each. 120 faded dots never answered and are still missing.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig3_what_weighting_fixes.png)

*Figure 3. An example phone poll of 1,001 people in a state Senate district. Weighting swaps the poll's age mix for the district's (left). It cannot bring back the people who never answered (right).*

**What post-stratification cannot do.** It fixes imbalances on the variables you weight by. It does not fix imbalances *within* those groups. Take a state Senate district where 18% of voters are under 30, and a phone poll of 1,001 people there (Figure 3). A fair poll of 1,000 would include about 180 voters under 30; the phone poll reached 60. Weighting counts each of them three times; it does not find the 120 who are missing. If the 60 who picked up are the politically engaged young voters, weighting makes the engaged three times louder, and the disengaged are still missing.

**Why not weight on everything?** First, you need a population target, and there is no census of, say, "trusts institutions". Second, groups get small: split the 60 young respondents by education and then by four regions and you have about seven people per group. Third, small groups produce extreme weights, so one respondent can move the result. Which variables a pollster weights on matters a lot: good pollsters handed the same raw interviews can publish results several points apart.

**Weighting costs precision.** Unequal weights make an estimate noisier. The **effective sample size** is the size of an unweighted poll that would be just as precise:

**n_eff = (sum of the weights)² ÷ (sum of the squared weights)**

If every weight is 1, n_eff equals the number of respondents. The more unequal the weights, the smaller n_eff and the wider the margin of error.

### The 2024 Cooperative Election Study

The **Cooperative Election Study (CES)** is a large survey of American adults run online, with people recruited from **online panels** (people who signed up in advance to take surveys on the internet). Our file has the **50,117** respondents in its fall 2024 pre-election wave who said they would definitely vote or had already voted. Think of it as one very large online poll of likely voters, taken in the weeks before the election. We know how that election came out: Trump won the national popular vote by about **1.5 points** (49.8% to 48.3%).

Each row is one **respondent**.

| Column | What it records | Values |
|---|---|---|
| `respondent_id` | Row number | 1, 2, 3, … |
| `age_group` | Age group | `18-29`, `30-44`, `45-64`, `65+` |
| `education` | Highest education | `HS or less`, `Some college`, `College degree` |
| `vote_2020` | Who they say they voted for in 2020 | `Biden`, `Trump`, `Other`, `Did not vote` |
| `vote_intent` | Who they planned to vote for (or had voted for) in 2024 | `dem` (Harris), `rep` (Trump), `other`, `undecided` |
| `ces_weight` | The survey's own full weight (see below) | positive numbers, average about 1 |

**Source:** Cooperative Election Study, Common Content 2024, Harvard Dataverse, [doi:10.7910/DVN/X11EP6](https://doi.org/10.7910/DVN/X11EP6), CC0. The [data notes](https://github.com/albertostefanelli/DSPC_coding_sessions/blob/master/wk08_trusting_polls/data/README.md) give the exact steps we used to build this file.

To weight a poll we need **targets**: what the electorate looked like. `ces_weight` is the survey's own full weight. The CES team built it to match Census figures on age, gender, education and race, then adjusted it further on voter registration, state, and **2020 and 2024 presidential vote**, among other things. The targets in the setup cell are the age, education and 2020-vote mix those weights imply for these likely voters.

```python
# Import the tools we need: pandas for tables, numpy for numbers, matplotlib for plots.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read the 2024 CES likely voters straight from the course repository (you need to be online).
# Source: Cooperative Election Study, Common Content 2024, doi:10.7910/DVN/X11EP6 (CC0).
# Python joins the two quoted strings inside the parentheses into one address.
ces = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/ces_2024_likely_voters.csv'
)

# Targets: the electorate's mix on age, education and 2020 vote (each set of shares sums to 1).
# A pandas Series is a column with labels; here the labels are the groups.
age_targets = pd.Series({'18-29': 0.130, '30-44': 0.229, '45-64': 0.345, '65+': 0.296})
education_targets = pd.Series({'HS or less': 0.272, 'Some college': 0.312, 'College degree': 0.416})
vote2020_targets = pd.Series({'Biden': 0.460, 'Trump': 0.429, 'Did not vote': 0.094, 'Other': 0.017})

# The actual result: Trump won the national popular vote by about 1.5 points.
actual_lead = -0.015

print('Respondents:', len(ces))
ces.head()
```

### 2a. Who responded?

Start with the toplines. A poll's **toplines** are its headline numbers: each candidate's share and the **lead**, one candidate's share minus the other's. As in Part 1, we score each respondent **+1** for Harris (`dem`), **−1** for Trump (`rep`) and **0** otherwise; the mean of the score is the lead.

**Units.** Shares are percentages (54.8%). Leads and differences between shares are **percentage points** (pp). Python gives proportions (0.548); we multiply by 100 when printing. A Democratic lead is positive and a Republican lead negative, so "Trump +1.5" is −1.5.

```python
# (ces['vote_intent'] == 'dem') is True for each Harris voter and False otherwise.
# .astype(int) turns True/False into 1/0, so the mean of the column is Harris's share.
ces['is_dem'] = (ces['vote_intent'] == 'dem').astype(int)
ces['is_rep'] = (ces['vote_intent'] == 'rep').astype(int)

# The score: +1 Harris, -1 Trump, 0 other or undecided. Its mean is the lead.
ces['score'] = ces['is_dem'] - ces['is_rep']
ces_lead = ces['score'].mean()

# Sampling error, as in Part 1: SE = spread of the score / square root of n.
se_lead = ces['score'].std() / np.sqrt(len(ces))

# Inside {}, :.1f means one decimal place and :+.1f also shows the sign (+ or -).
print(f'Harris {100*ces["is_dem"].mean():.1f}%   Trump {100*ces["is_rep"].mean():.1f}%')
print(f'Lead:            {100*ces_lead:+.1f} pp')
print(f'Margin of error: ±{100*1.96*se_lead:.1f} pp  '
      f'(95% CI {100*(ces_lead - 1.96*se_lead):+.1f} to {100*(ces_lead + 1.96*se_lead):+.1f})')
print(f'Actual result:   {100*actual_lead:+.1f} pp')
```

The raw sample has Harris ahead by **15.3 points**, with a margin of error of less than **±1 point**. Trump won by 1.5. The sample misses by almost **17 points**, about twenty times its own margin of error. With 50,000 respondents, sampling error is tiny, so this is the bottom-right panel of Figure 1: a very precise wrong answer. The problem is *who answered*. Start by comparing the sample's age mix with the electorate's.

```python
# Share of respondents in each age group (shares, not counts).
ces_age = ces['age_group'].value_counts(normalize=True)

# pd.DataFrame matches rows by label, so every 18-29 share sits on the same row.
age_comparison = pd.DataFrame({'Electorate (target)': age_targets, 'CES sample': ces_age})
(100 * age_comparison).round(1)
```

```python
# A grouped bar chart of the table above: one pair of bars per age group.
ax = (100 * age_comparison).plot.bar(figsize=(7, 4), rot=0, width=0.75,
                                     color=['#2c3e50', '#3498db'], edgecolor='white')
ax.set_xlabel('Age group')
ax.set_ylabel('Share of likely voters (%)')
ax.set_title('Who responded? The CES sample against the electorate')
plt.tight_layout()
plt.show()
```

```python
# The lead inside each age group (groupby, as in Week 2).
lead_by_age = ces.groupby('age_group')['score'].mean()
(100 * lead_by_age).round(1)
```

The sample has too few people aged 30–44 (17% against 23%) and somewhat too many aged 45 and over. Now look at how each group votes: respondents under 45 favor Harris by about 30 points or more, those 45 and over by about 7 to 10. So the age mix is off, but in a direction that should make the raw sample lean *toward Trump*, not Harris. Whatever makes this sample 17 points too pro-Harris, it is not its age mix. Keep that prediction in mind for the next step.

### 2b. Post-stratification: reweight by age

**weight = target share ÷ sample share.** Dividing one Series by another with the same labels divides the 18–29 target by the 18–29 sample share, the 65+ target by the 65+ share, and so on. Pandas matches the rows by label for you.

```python
# One weight per age group: electorate share / sample share.
age_weights = age_targets / ces_age
age_weights.round(2)
```

Each respondent aged 30–44 now counts as 1.32 people, and each respondent aged 45 or over as about 0.9 of a person. To get the weighted lead, average the lead in each age group, letting each group count for its share of the sample **times its weight**. Since sample share × weight = target share, this swaps the sample's age mix for the electorate's.

```python
# Weighted: multiply each group's sample share by its weight first.
# (sample share x weight = target share, so this is the electorate's age mix.)
age_weighted_lead = (ces_age * age_weights * lead_by_age).sum()

print(f'Unweighted lead:      {100*ces_lead:+.1f} pp')
print(f'Weighted by age:      {100*age_weighted_lead:+.1f} pp')
print(f'Actual result:        {100*actual_lead:+.1f} pp')
```

```python
# Plot the two estimates against the actual result (dashed line).
labels = ['Unweighted', 'Weighted\nby age']
leads = [100*ces_lead, 100*age_weighted_lead]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(labels, leads, color=['#95a5a6', '#3498db'], edgecolor='black', linewidth=0.5, width=0.5)
ax.axhline(0, color='black', linewidth=1)
ax.axhline(100*actual_lead, color='#e74c3c', linestyle='--', linewidth=1.5, label='Actual result: Trump +1.5')
for j, v in enumerate(leads):
    ax.text(j, v + 0.4, f'{v:+.1f}', ha='center', fontsize=11)
ax.set_ylabel('Lead (pp, positive = Harris ahead)')
ax.set_title('Weighting by age moves the estimate the wrong way')
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

Weighting by age moved the lead from Harris +15.3 to **Harris +16.8**, *further* from the result, as 2a predicted. Nothing went wrong with the arithmetic: the weighted sample now has exactly the right age mix. But fixing the age mix gave more weight to people aged 30–44, who lean toward Harris, and the sample's real problem lies somewhere else.

This is the limit from Part 1 in real data. Post-stratification fixes imbalances on the variables you weight by. It cannot fix imbalances on variables you leave out, or differences between the people who answer and the people who don't *within* each group. Exercise 1 asks which variable would actually help.

Weighting also has a cost. To measure it, give every respondent the weight of their age group, then compute the effective sample size.

```python
# .map() looks up each respondent's age group in age_weights and returns that group's weight.
ces['age_weight'] = ces['age_group'].map(age_weights)

# Check: the weighted mean, sum(answer x weight) / sum(weight), gives the same lead as above.
check = (ces['score'] * ces['age_weight']).sum() / ces['age_weight'].sum()
print(f'Weighted lead, row by row: {100*check:+.1f} pp')

# Effective sample size: (sum of weights)^2 / (sum of squared weights).
# :, adds a thousands separator (50,117).
neff_age = ces['age_weight'].sum()**2 / (ces['age_weight']**2).sum()
print(f'{len(ces):,} respondents, effective sample size {neff_age:,.0f}')
```

The age weights are mild (0.90 to 1.32), so weighting costs little precision: 50,117 respondents become an effective **48,895**. Unequal weights always cost something, and extreme weights cost much more: a group weighted ×3, like the young voters in Figure 3, shrinks a poll far faster.

### Setup for Exercises 2–4: the FiveThirtyEight poll archive

How often do real polls miss, and by how much? **The FiveThirtyEight poll archive** has **8,589 real polls** taken in the last 21 days before Democrat-versus-Republican general elections for president, Senate, governor and House, from 1998 to 2022. Each row is one **poll**, with the lead it reported and the actual result.

**Source:** FiveThirtyEight collected these polls to rate pollsters and published them as [`pollster-ratings/raw_polls.csv`](https://github.com/fivethirtyeight/data/blob/be387bc1649413be1c85f2181b2c32aacee24315/pollster-ratings/raw_polls.csv) (January 2024 version, CC BY 4.0). Our copy keeps only the races and columns described here and renames a few columns; the [data notes](https://github.com/albertostefanelli/DSPC_coding_sessions/blob/master/wk08_trusting_polls/data/README.md) list every step.

| Column | What it records |
|---|---|
| `cycle` | Election year |
| `race`, `location`, `race_type` | Which race (for example `2016_Pres-G_US`), the state or `US`, and the office |
| `pollster`, `methodology` | Who ran the poll, and its **mode**, meaning how people were contacted (`Live Phone`, `Online Panel`, `IVR` = automated phone, …) |
| `sponsor` | Who paid for it: `Nonpartisan`, `Democratic` or `Republican` (a party or campaign) |
| `poll_date`, `sample_size` | When it was taken and how many people it interviewed |
| `margin_poll` | The poll's lead, Democrat minus Republican, in points |
| `margin_actual` | The actual result, Democrat minus Republican, in points |

The setup cell keeps only races decided by **fewer than 10 points**. Those are the races where polls matter most, and there is a technical reason too: `margin_poll` comes from shares that still include undecided voters, so in a lopsided race a poll's lead looks smaller than the final margin even if the undecided split like everyone else.

The archive does not give each poll's shares, so the setup cell computes each poll's margin of error on the lead with p = 0.5. That gives the widest possible margin for a simple random sample, which is generous to the polls. If sampling error were the only error, at most about **1 in 20** polls would miss the result by more than this margin.

```python
# Read the FiveThirtyEight poll archive from the course repository.
# Source: FiveThirtyEight, pollster-ratings/raw_polls.csv (January 2024 version, CC BY 4.0),
# filtered to the last 21 days of Dem-vs-Rep general elections, 1998-2022.
polls = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/final_polls_1998_2022.csv'
)

# Keep races decided by fewer than 10 points (either party). .abs() drops the minus sign.
polls = polls[polls['margin_actual'].abs() < 10].copy()

# How far each poll's lead was from the actual result.
# Positive error = the poll was too good for the Democrat; negative = too good for the Republican.
polls['error'] = polls['margin_poll'] - polls['margin_actual']
polls['abs_error'] = polls['error'].abs()          # size of the miss, ignoring direction

# Each poll's own margin of error on the lead, in points.
# With p = 0.5 a share has SE sqrt(0.25 / n) = 0.5 / sqrt(n); a lead has about twice that (Part 1).
polls['moe_lead'] = 1.96 * 2 * np.sqrt(0.25 / polls['sample_size']) * 100

print('Close-race polls:', len(polls))
polls.head()
```
