## 1. Why do polls of the same race disagree?

### The story

Eight weeks before Election Day, three polls of the same state Senate race are released:

| Poll | How people were contacted | Who was surveyed | Sample size | Result |
|---|---|---|---|---|
| Landmark Research | Live phone calls | Likely voters | about 1,000 | Rep +4 |
| Progressive Data Institute | Online panel | Likely voters | about 1,000 | Dem +8 |
| State University Poll | Phone and online, then weighted | Registered voters | about 1,200 | Dem +1 |

An **online panel** is a group of people who signed up in advance to take surveys on the internet. Every number in the last column is a **lead**: one candidate's share of respondents minus the other's. A poll's headline numbers (each candidate's share and the lead) are called its **toplines**. The phone and online polls are 12 points apart.

**Units.** Shares are percentages (46.2%). Leads and differences between shares are **percentage points** (pp, or just "points"). Python gives you proportions (0.462); multiply by 100 to report either. We write a Democratic lead as a positive number and a Republican lead as a negative one, so "Rep +4" is −4. Before deciding which one to trust, we need to know how far apart two honest polls of the same race could land just by chance, and what else could be going on.

### Five reasons two polls disagree

Three reasons come from **collecting** the data:

- **Sampling error.** A poll interviews about 1,000 people to learn what hundreds of thousands think. Interview a different 1,000 and you get a different number. This is random noise.
- **Nonresponse bias.** The people who answer the survey differ systematically from the people who don't, in ways that affect the answer. If Republican-leaning voters are less likely to pick up the phone, the poll overstates Democratic support, even if the pollster dialed exactly the right mix of numbers. Phone response rates are now below 5%, so the people who do answer are not a random sample. Phone polls tend to reach older voters, who answer calls from unknown numbers. Online panels tend to reach younger, more politically engaged people.
- **Mode effects.** The **mode** is how people are contacted: a live phone call, an automated call, an online panel, a text message with a link. A **mode effect** means the same person gives a different answer depending on the mode, for example saying what sounds good when a live interviewer is listening.

Two more come from **analyzing** the data:

- **Weighting.** Which groups the pollster adjusts for, and to what targets.
- **Likely-voter screen.** The pollster's rule for deciding which respondents count as voters: asking "Do you plan to vote?", checking past turnout in the voter file, or a mix of both. Different rules produce different electorates from the same interviews.

Both are **decisions the pollster makes**. They rely on measured information (survey answers, Census figures, the voter file), but the pollster chooses which information to use and how to combine it. Two careful pollsters can make different reasonable choices with the same interviews and publish different numbers.

Only the first reason is random. The other four can push a poll in one direction.

### Two kinds of error: noise and bias

**Sampling error** is noise. It pushes a poll up or down at random, and it shrinks as the sample grows. **Bias** pushes every poll from the same method in the same direction, and it does **not** shrink when you collect more of the same kind of data. A poll of 50,000 biased respondents is a more precise wrong answer.

![Four panels of 200 simulated polls each. Top row, an unbiased pollster: polls cluster around the true Democratic share of 48 percent, widely with 500 respondents and tightly with 5,000. Bottom row, a pollster whose respondents lean Democratic: polls cluster around 51 percent, and with 5,000 respondents they cluster tightly around the wrong number.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig1_noise_vs_bias.png)

*Figure 1. Each dot is one simulated poll. A bigger sample tightens the cloud (left to right). It does not move the cloud back to the truth (bottom row).*

The **margin of error** printed next to a poll measures only the first kind of error. You can compute it from the poll itself. You cannot compute the bias from the poll; you can only argue about which direction it goes. In 2020, national presidential polls overstated Biden's margin by 3.9 points, and nearly all of them missed in the same direction. Averaging more of those polls would not have helped.

Bias can also come from the pollster. A **house effect** is a pollster's methods (how it contacts people, words questions, weights and screens for likely voters) consistently favoring one side. Polls paid for by a party or campaign deserve extra care for a second reason: the campaign decides which of its internal polls to release, and it tends to release the ones it likes. Picking the poll that tells you what you want to hear is **cherry-picking**, and it is how a set of honest polls ends up telling a dishonest story. Before trusting a poll, ask who paid for it, who was interviewed, how they were contacted, and how the results were weighted.

### Measuring sampling error: the margin of error

The margin of error is the "plus or minus" of a 95% confidence interval, the same idea as in Week 5, applied to a poll number instead of a treatment effect:

**margin of error = 1.96 × standard error**

**standard error = (how far a typical respondent is from the average) ÷ √n**

The bottom of the fraction is the sample size: four times the sample gives half the standard error. The top is how spread out people are on the question. For a yes/no question such as "Did you pick the Democrat?", the top is √(p × (1 − p)), where p is the share who say yes. In the phone poll, 46% of 1,001 respondents picked the Democrat:

> √(0.46 × 0.54) ÷ √1,001 = 0.50 ÷ 31.6 ≈ 0.016, or 1.6 points. The margin of error is 1.96 × 1.6 ≈ ±3 points, so "Dem 46%" means "Dem somewhere between about 43% and 49%".

**A lead is noisier than a share.** If one respondent switches from the Democrat to the Republican, the Democrat's share falls by one person and the Republican's rises by one, so the lead moves by two. To measure a lead, score each respondent **+1** for the Democrat, **−1** for the Republican and **0** for undecided. The average score is the lead. Almost everyone picks a side, so a typical score sits about 1 away from the average, against about 0.5 for a yes/no answer. Same sample size, twice the spread, twice the standard error:

| Question | Possible answers | Typical distance from the average | Standard error | Margin of error |
|---|---|---|---|---|
| "Did you pick the Democrat?" (a share) | 0 or 1 | about 0.5 | 1.6 points | ±3 points |
| "Which way do you lean?" (a lead) | −1, 0 or +1 | about 1.0 | 3.1 points | ±6 points |

![Left: the phone poll's Democratic share of 46 percent with a margin of error of plus or minus 3 points. Right: three leads with 95 percent intervals on an axis from Rep +12 to Dem +16. The phone poll, Rep +4, and the university poll, Dem +1, both cross the tie line and are statistical ties. The online poll, Dem +8, clears the tie line but its interval reaches down to about Dem +2.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig2_share_vs_lead.png)

*Figure 2. The same sample size gives a share a margin of about ±3 points and a lead a margin of about ±6. Two of the three leads are statistical ties.*

A lead whose 95% interval includes zero is a **statistical tie**: the poll cannot tell us who is ahead. This is the Week 5 significance test again: is the lead distinguishable from zero?

**Comparing two polls.** The difference between two independent polls has its own standard error: square each poll's standard error, add them, and take the square root. With 3.1 points each, √(3.1² + 3.1²) ≈ 4.4 points. So two polls of 1,000 people need to be about 1.96 × 4.4 ≈ 9 points apart before we can rule out chance as the whole explanation.

### Fixing what you can: post-stratification

**Post-stratification** means adjusting the sample after the interviews are done so that it matches the population on characteristics we know from the census or the voter file, such as age, race, education and party registration. If the sample has too many older people, each younger respondent counts for more.

1. For each group, compute a **weight = population share of the group ÷ sample share of the group**.
2. Recompute the results, counting each respondent as many times as their weight.

A **weighted mean** is the sum of (answer × weight) divided by the sum of the weights.

> Two respondents: one picked the Democrat and has weight 3, the other picked the Republican and has weight 1. The weighted Democratic share is (1 × 3 + 0 × 1) ÷ (3 + 1) = 3/4 = 75%. Unweighted, it would be 1/2 = 50%.

![Left: the phone poll's age mix next to the district's. The poll has 6 percent aged 18 to 29 against 18 percent in the district, and 41 percent aged 65 and over against 25 percent. The weights are 3.00, 1.47, 0.90 and 0.60. Right: 180 dots for the young voters a fair poll of 1,000 would include. 60 filled dots answered and now count three times each. 120 faded dots never answered and are still missing.](https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/refs/heads/master/wk08_trusting_polls/assets/fig3_what_weighting_fixes.png)

*Figure 3. Weighting swaps the poll's age mix for the district's (left). It cannot bring back the people who never answered (right).*

**What post-stratification cannot do.** It fixes imbalances on the variables you weight by. It does not fix imbalances *within* those groups. A fair poll of 1,000 in our district would include about 180 voters under 30. The phone poll reached 60. Weighting counts each of them three times; it does not find the 120 who are missing. If the 60 who picked up are the politically engaged young voters, weighting makes the engaged three times louder, and the disengaged are still missing.

**Why not weight on everything?** First, you need a population target, and there is no census of, say, "trusts institutions". Second, groups get small: split the 60 young respondents by education and then by four regions and you have about seven people per group. Third, small groups produce extreme weights, so one respondent can move the result. Which variables a pollster weights on matters a lot: good pollsters handed the same raw interviews can publish results several points apart.

**Weighting costs precision.** Unequal weights make an estimate noisier. The **effective sample size** is the size of an unweighted poll that would be just as precise:

**n_eff = (sum of the weights)² ÷ (sum of the squared weights)**

If every weight is 1, n_eff equals the number of respondents. The more unequal the weights, the smaller n_eff and the wider the margin of error.

### Two polls of the same race

We have the raw interviews behind two of the three polls in the story: the phone poll and the online poll. Remember that these are **teaching data**. The university poll's raw data are not available, so we only use its published result, Dem +1.

Each row is one **respondent**.

| Column | What it records | Values |
|---|---|---|
| `respondent_id` | Respondent number | 1, 2, 3, … |
| `age_group` | The respondent's age group | `18-29`, `30-44`, `45-64`, `65+` |
| `vote_intent` | Who the respondent plans to vote for | `dem`, `rep`, `undecided` |
| `poll` | Which poll the respondent is from | `phone_poll`, `online_poll` |

We also need the population we want the polls to describe. The district's voter file gives the age mix of all registered voters: 18.0% are 18–29, 25.0% are 30–44, 32.1% are 45–64 and 24.9% are 65 or older.

```python
# Import the tools we need: pandas for tables, numpy for numbers, matplotlib for plots.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read both polls straight from the course repository (you need to be online).
# Source: teaching data built for this course (see the data notes linked above).
# Python joins the two quoted strings inside the parentheses into one address.
phone = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/phone_poll.csv'
)
online = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/online_poll.csv'
)

# Age mix of ALL registered voters in the district, from the voter file.
# A pandas Series is a column with labels; here the labels are the age groups.
pop_age = pd.Series({'18-29': 0.180, '30-44': 0.250, '45-64': 0.321, '65+': 0.249})

print('Phone poll respondents: ', len(phone))
print('Online poll respondents:', len(online))
phone.head()
```

### 2a. Toplines: shares and leads

A column of words cannot be averaged, so we first turn `vote_intent` into numbers. `is_dem` is 1 for a Democratic respondent and 0 otherwise; its mean is the Democratic share. `score` is +1, −1 or 0; its mean is the lead.

```python
# Phone poll.
# (phone['vote_intent'] == 'dem') is True for each Democratic respondent and False otherwise.
# .astype(int) turns True/False into 1/0, so the mean of the column is the Dem share.
phone['is_dem'] = (phone['vote_intent'] == 'dem').astype(int)
phone['is_rep'] = (phone['vote_intent'] == 'rep').astype(int)

# The score: +1 for Dem, -1 for Rep, 0 for undecided (1 - 0, 0 - 1, or 0 - 0).
# Its mean is the lead: Dem share minus Rep share.
phone['score'] = phone['is_dem'] - phone['is_rep']

# Online poll: the same three columns.
online['is_dem'] = (online['vote_intent'] == 'dem').astype(int)
online['is_rep'] = (online['vote_intent'] == 'rep').astype(int)
online['score'] = online['is_dem'] - online['is_rep']

# Multiply proportions by 100 only when printing.
# Inside {}, :.1f means one decimal place and :+.1f also shows the sign (+ or -).
print(f'Phone poll:  Dem {100*phone["is_dem"].mean():.1f}%   Rep {100*phone["is_rep"].mean():.1f}%   '
      f'Lead {100*phone["score"].mean():+.1f} pp')
print(f'Online poll: Dem {100*online["is_dem"].mean():.1f}%   Rep {100*online["is_rep"].mean():.1f}%   '
      f'Lead {100*online["score"].mean():+.1f} pp')
```

The phone poll has the Republican ahead by about 4 points (a lead of −3.9); the online poll has the Democrat ahead by about 8 (+7.6). The lead is computed from unrounded shares, so it can differ by 0.1 from subtracting the rounded shares you see printed.

### 2b. Sampling error: a real lead or a statistical tie?

The standard error is the spread of the answers divided by √n. `.std()` computes the top of that fraction: how far a typical respondent sits from the average. We compute it twice for the phone poll, once for the 0/1 Democratic column (a share) and once for the +1/0/−1 score (a lead).

```python
n_phone = len(phone)

# Spread of each column: how far a typical respondent is from the average.
spread_share = phone['is_dem'].std()
spread_lead = phone['score'].std()

# Standard error = spread / square root of the sample size.
se_share_phone = spread_share / np.sqrt(n_phone)
se_lead_phone = spread_lead / np.sqrt(n_phone)

print(f'Dem share: spread {spread_share:.2f}, SE {100*se_share_phone:.1f} pp, '
      f'margin of error ±{100*1.96*se_share_phone:.1f} pp')
print(f'Lead:      spread {spread_lead:.2f}, SE {100*se_lead_phone:.1f} pp, '
      f'margin of error ±{100*1.96*se_lead_phone:.1f} pp')
```

Same 1,001 people, two different margins of error. The ±3 printed under a poll describes **one candidate's share**. The lead, which is what campaigns and reporters talk about, is about twice as uncertain: ±6 points. Now the 95% interval for each poll's lead.

```python
# Each poll's lead and its standard error.
lead_phone = phone['score'].mean()
lead_online = online['score'].mean()
se_lead_online = online['score'].std() / np.sqrt(len(online))

# 95% confidence interval: the lead plus or minus 1.96 standard errors.
print(f'Phone:  lead {100*lead_phone:+.1f} pp, 95% CI {100*(lead_phone - 1.96*se_lead_phone):+.1f} '
      f'to {100*(lead_phone + 1.96*se_lead_phone):+.1f}')
print(f'Online: lead {100*lead_online:+.1f} pp, 95% CI {100*(lead_online - 1.96*se_lead_online):+.1f} '
      f'to {100*(lead_online + 1.96*se_lead_online):+.1f}')
```

The phone poll's interval runs from about Rep +10 to Dem +2. It includes zero, so it is a **statistical tie**: the poll leans Republican but cannot tell us who is ahead. The online poll's interval clears zero, but only just: it is consistent with a 2-point race. Next, is the gap *between* the two polls bigger than chance alone would produce?

```python
# The gap between the two polls has its own standard error:
# square each poll's SE, add them, take the square root. ** means "to the power of".
gap = lead_online - lead_phone
se_gap = np.sqrt(se_lead_phone**2 + se_lead_online**2)

print(f'Gap:    {100*gap:.1f} pp')
print(f'SE:     {100*se_gap:.1f} pp')
print(f'95% CI: {100*(gap - 1.96*se_gap):+.1f} to {100*(gap + 1.96*se_gap):+.1f} pp')
```

The gap is 11.5 points. Headlines say 12 because they subtract the rounded toplines, 8 − (−4). The interval runs from about 3 to 20 points and excludes zero, so the two polls disagree by more than sampling error alone would produce. Be careful about what this does and does not show. It tells us that something other than chance separates the two polls. It does **not** tell us which poll is closer to the truth. Interviewing more people the same way would only shrink both intervals around roughly the same two numbers.

### 2c. Who responded?

If chance does not explain the gap, look at *who* each poll reached. `value_counts(normalize=True)` gives the share of respondents in each age group. Putting the three Series into one DataFrame lines them up by their age-group labels.

```python
# Share of each poll's respondents in each age group (shares, not counts).
phone_age = phone['age_group'].value_counts(normalize=True)
online_age = online['age_group'].value_counts(normalize=True)

# pd.DataFrame matches rows by label, so every 18-29 share sits on the same row.
comparison = pd.DataFrame({
    'District (voter file)': pop_age,
    'Phone poll': phone_age,
    'Online poll': online_age,
})
(100 * comparison).round(1)
```

```python
# A grouped bar chart of the table above: one group of bars per age group.
ax = (100 * comparison).plot.bar(figsize=(8, 4), rot=0, width=0.8,
                                 color=['#2c3e50', '#95a5a6', '#3498db'], edgecolor='white')
ax.set_xlabel('Age group')
ax.set_ylabel('Share of respondents (%)')
ax.set_title('Each poll over-represents the people most likely to answer it')
plt.tight_layout()
plt.show()
```

```python
# The lead inside each age group, in each poll (groupby, as in Week 2).
lead_by_age = pd.DataFrame({
    'Phone poll': phone.groupby('age_group')['score'].mean(),
    'Online poll': online.groupby('age_group')['score'].mean(),
})
(100 * lead_by_age).round(1)
```

The phone poll is 41% over 65, against 25% of registered voters. The online poll is 33% under 30, against 18%. Each poll over-represents the age group most likely to answer it.

Now look at the second table. **Inside each age group the two polls agree closely**: young respondents favor the Democrat by about 25 points in both polls, and respondents over 65 favor the Republican by 11 to 13 points in both. So the polls disagree mainly because they reached different *mixes* of people, not because similar people answered differently. A phone poll full of older voters leans Republican; an online poll full of younger voters leans Democratic.

### 2d. Post-stratification: reweight both polls by age

**weight = population share ÷ sample share.** Dividing one Series by another with the same labels divides the 18–29 share by the 18–29 share, the 65+ share by the 65+ share, and so on. Pandas matches the rows by label for you.

```python
# One weight per age group: district share / sample share.
phone_weights = pop_age / phone_age
online_weights = pop_age / online_age

pd.DataFrame({'Phone weight': phone_weights, 'Online weight': online_weights}).round(2)
```

Each young phone respondent now counts as three people, and each phone respondent over 65 counts as 0.6 of a person. The online poll needs the opposite fix: its respondents over 65 count 2.19 times each.

To get the weighted lead, average the lead in each age group, letting each group count for its share of the sample **times its weight**. Since sample share × weight = district share, this swaps the poll's age mix for the district's.

```python
# Lead in each age group (from 2c).
phone_lead_by_age = phone.groupby('age_group')['score'].mean()
online_lead_by_age = online.groupby('age_group')['score'].mean()

# Weighted: multiply each group's sample share by its weight first.
# (sample share x weight = district share, so this is the district's age mix.)
phone_weighted_lead = (phone_age * phone_weights * phone_lead_by_age).sum()
online_weighted_lead = (online_age * online_weights * online_lead_by_age).sum()

print(f'Phone lead:  {100*lead_phone:+.1f} -> {100*phone_weighted_lead:+.1f} pp')
print(f'Online lead: {100*lead_online:+.1f} -> {100*online_weighted_lead:+.1f} pp')
print(f'Gap:         {100*gap:.1f} -> {100*(online_weighted_lead - phone_weighted_lead):.1f} pp')
```

```python
# Plot the leads before and after weighting, with the university poll (Dem +1) as a dashed line.
labels = ['Phone\nunweighted', 'Phone\nweighted', 'Online\nunweighted', 'Online\nweighted']
leads = [100*lead_phone, 100*phone_weighted_lead, 100*lead_online, 100*online_weighted_lead]
colors = ['#95a5a6', '#2c3e50', '#d6eaf8', '#3498db']

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(labels, leads, color=colors, edgecolor='black', linewidth=0.5, width=0.6)
ax.axhline(0, color='black', linewidth=1)
ax.axhline(1, color='#e74c3c', linestyle='--', linewidth=1.5, label='University poll: Dem +1')
for j, v in enumerate(leads):
    ax.text(j, v + (0.4 if v >= 0 else -1.0), f'{v:+.1f}', ha='center', fontsize=11)
ax.set_ylabel('Lead (pp, positive = Democrat ahead)')
ax.set_title('Weighting by age brings both polls close to Dem +1')
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

Weighting by age moves the phone poll from Rep +3.9 to Dem +1.9 and the online poll from Dem +7.6 to Dem +1.6. The 11.5-point gap is gone, and both polls land near the university poll's Dem +1.

That is tidier than real life. These data were built so that age explains the disagreement. In real polls, weighting usually closes some of the gap and leaves the rest, because the people who answer differ from the people who don't *within* every age group, and nobody observes that. The weighted numbers are only as good as the assumption that, inside each age group, the people who answered vote like the people who didn't.

Weighting also has a cost. To measure it, give every respondent the weight of their age group, then compute the effective sample size.

```python
# .map() looks up each respondent's age group in phone_weights and returns that group's weight.
phone['weight'] = phone['age_group'].map(phone_weights)
online['weight'] = online['age_group'].map(online_weights)

# Check: the weighted mean, sum(answer x weight) / sum(weight), gives the same lead as above.
check = (phone['score'] * phone['weight']).sum() / phone['weight'].sum()
print(f'Phone weighted lead, row by row: {100*check:+.1f} pp')

# Effective sample size: (sum of weights)^2 / (sum of squared weights).
neff_phone = phone['weight'].sum()**2 / (phone['weight']**2).sum()
neff_online = online['weight'].sum()**2 / (online['weight']**2).sum()
# :, adds a thousands separator (1,001).
print(f'Phone:  {len(phone):,} respondents, effective sample size {neff_phone:.0f}')
print(f'Online: {len(online):,} respondents, effective sample size {neff_online:.0f}')

# Margin of error on the phone poll's Dem share, before and after weighting.
dem_unweighted = phone['is_dem'].mean()
dem_weighted = (phone['is_dem'] * phone['weight']).sum() / phone['weight'].sum()
moe_before = 1.96 * np.sqrt(dem_unweighted * (1 - dem_unweighted) / len(phone))
moe_after = 1.96 * np.sqrt(dem_weighted * (1 - dem_weighted) / neff_phone)
print(f'Phone Dem share: {100*dem_unweighted:.1f}% ± {100*moe_before:.1f} unweighted, '
      f'{100*dem_weighted:.1f}% ± {100*moe_after:.1f} weighted')
```

Weighting moved the phone poll's Democratic share by 2.4 points and cost about 258 interviews' worth of precision: the 1,001 weighted respondents are only as precise as 743 unweighted ones, so the margin of error on the share grows from ±3.1 to ±3.6. The online poll's weights are less extreme, so it loses less. Weighting trades a smaller bias on age for more noise.

### 2e. Real polls: how often do they miss by more than the margin of error?

The two polls above are teaching data. To see how real polls do, we use **the FiveThirtyEight poll archive**: **8,589 real polls** taken in the last 21 days before Democrat-versus-Republican general elections for president, Senate, governor and House, from 1998 to 2022. Each row is one **poll**, with the lead it reported and the actual result.

**Source:** FiveThirtyEight collected these polls to rate pollsters and published them as [`pollster-ratings/raw_polls.csv`](https://github.com/fivethirtyeight/data/blob/be387bc1649413be1c85f2181b2c32aacee24315/pollster-ratings/raw_polls.csv) (January 2024 version, CC BY 4.0). Our copy keeps only the races and columns described here and renames a few columns; the [data notes](https://github.com/albertostefanelli/DSPC_coding_sessions/blob/master/wk08_trusting_polls/data/README.md) list every step.

| Column | What it records |
|---|---|
| `cycle` | Election year |
| `race`, `location`, `race_type` | Which race (for example `2016_Pres-G_US`), the state or `US`, and the office |
| `pollster`, `methodology` | Who ran the poll, and the mode (`Live Phone`, `Online Panel`, `IVR` = automated phone, …) |
| `sponsor` | Who paid for it: `Nonpartisan`, `Democratic` or `Republican` (a party or campaign) |
| `poll_date`, `sample_size` | When it was taken and how many people it interviewed |
| `margin_poll` | The poll's lead, Democrat minus Republican, in points |
| `margin_actual` | The actual result, Democrat minus Republican, in points |

We keep only races decided by **fewer than 10 points**. Those are the races where polls matter most, and there is a technical reason too: `margin_poll` comes from shares that still include undecided voters, so in a lopsided race a poll's lead looks smaller than the final margin even if the undecided split like everyone else. Close races keep that problem small.

The FiveThirtyEight archive does not give each poll's shares, so we compute each poll's margin of error on the lead with p = 0.5. That gives the widest possible margin for a simple random sample, which is generous to the polls. If sampling error were the only error, at most about **1 in 20** polls would miss the result by more than this margin.

```python
# Read the FiveThirtyEight poll archive from the course repository.
# Source: FiveThirtyEight, pollster-ratings/raw_polls.csv (January 2024 version, CC BY 4.0),
# filtered to the last 21 days of Dem-vs-Rep general elections, 1998-2022.
polls = pd.read_csv(
    'https://raw.githubusercontent.com/albertostefanelli/DSPC_coding_sessions/'
    'master/wk08_trusting_polls/data/final_polls_1998_2022.csv'
)
print('Rows and columns, all races:', polls.shape)

# Keep races decided by fewer than 10 points (either party). .abs() drops the minus sign.
polls = polls[polls['margin_actual'].abs() < 10].copy()
print('Rows and columns, close races:', polls.shape)
polls.head()
```

```python
# How far each poll's lead was from the actual result.
# Positive error = the poll was too good for the Democrat; negative = too good for the Republican.
polls['error'] = polls['margin_poll'] - polls['margin_actual']
polls['abs_error'] = polls['error'].abs()          # size of the miss, ignoring direction

# Each poll's own margin of error on the lead, in points.
# With p = 0.5 a share has SE sqrt(0.25 / n) = 0.5 / sqrt(n); a lead has about twice that (Part 1).
polls['moe_lead'] = 1.96 * 2 * np.sqrt(0.25 / polls['sample_size']) * 100

# True if the poll missed the result by more than its own margin of error.
polls['outside_moe'] = polls['abs_error'] > polls['moe_lead']

print(f'Typical sample size:                {polls["sample_size"].median():.0f}')
print(f'Average margin of error on the lead: ±{polls["moe_lead"].mean():.1f} pp')
print(f'Share of polls outside their own margin of error: {polls["outside_moe"].mean():.1%}')
```

```python
# Histogram of the errors. The shaded band is the margin of error of a typical poll.
typical_moe = polls['moe_lead'].median()

fig, ax = plt.subplots(figsize=(8, 4))
ax.axvspan(-typical_moe, typical_moe, color='#d6eaf8', label=f'Typical margin of error (±{typical_moe:.1f})')
ax.hist(polls['error'], bins=range(-30, 31), color='#2c3e50', edgecolor='white')
ax.axvline(0, color='#e74c3c', linewidth=1.5)
ax.set_xlabel('Poll lead minus actual result (pp)\n<- too good for the Republican | too good for the Democrat ->')
ax.set_ylabel('Number of polls')
ax.set_title('Close races: final polls miss in both directions, often by more than the margin')
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

About **1 in 6** final polls in close races missed the result by more than its own margin of error, several times the 1 in 20 that sampling error alone would produce. Part of the gap is weighting: as 2d showed, weighted polls have a somewhat wider margin than this formula. But weighting cannot explain all of it. The margin of error is not the total error: it leaves out nonresponse, mode effects, weighting and likely-voter choices, and voters who decide late.

That does not make polls useless. Most misses are a few points, so a poll showing a 15-point lead almost always picks the right winner. The trouble is close races like these: when the lead is smaller than the typical miss, the poll tells you little about who will win, which is exactly where campaigns need it most.
