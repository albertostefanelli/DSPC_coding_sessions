
## 2. Converting to and from strings (type casting)

In Python, every piece of data has a specific *type*. For instance, `5` is an integer (`int`), `3.14` is a floating-point number (`float`), and `"hello"` is a string (`str`). Understanding data types is fundamental because different types behave differently and support different operations. For example, you can perform arithmetic operations on numbers, but not directly on strings (e.g., `"5" + "2"` results in `"52"`, not `7`).

**Type casting** (or type conversion) is the process of converting a value from one data type to another. This is a very common operation, especially when dealing with user input (which is often read as a string) or when preparing data for specific operations.



### Check the type before converting: `type()`

Use the built-in `type(value)` to ask Python what kind of object you have. The result `<class 'str'>` means “this value is a string.” Here, *class* is Python's term for the object's type; you do not need to write your own class.

Numbers inside quotation marks are text. Inspecting a type does not change the value.


```python
print(type("25"))   # Text: str
print(type(25))     # Whole number: int
print(type(25.0))   # Decimal number: float
print(type(True))   # True/False value: bool
print(type([25]))   # Collection: list
```

### Converting to a string (`str()`)

The `str()` function is used to convert values of other data types into strings. This is particularly useful when you want to concatenate (join) numbers or other data with text, or when you need to store non-string data in a text-based format.

#### Converting numbers to strings

Let's say you want to display a message that includes a numerical value.



```python
age = 20
print("I am " + str(age) + " years old.")
```


### Converting from a string (`int()`, `float()`)

Conversely, you often receive data as strings (e.g., from user input, file reading, or web scraping) that you need to treat as numbers. Python provides `int()` for converting to integers and `float()` for converting to floating-point numbers.

Consider a scenario where a user inputs their age or a price.



```python
user_age_str = "25"
user_age_int = int(user_age_str)
print("User's age (as integer): " + str(user_age_int) + ", Type: " + str(type(user_age_int)))

item_price_str = "19.99"
item_price_float = float(item_price_str)
print("Item price (as float): " + str(item_price_float) + ", Type: " + str(type(item_price_float)))

# You can now perform arithmetic operations
total_cost = item_price_float * 2
print("Total cost for two items: " + str(total_cost))
```

## 4. Merging data

In real-world data analysis, information is rarely confined to a single table or file. Often, you'll encounter scenarios where relevant data is spread across multiple sources, and you need to combine them to gain a complete picture. The process of combining data based on common columns or keys is known as **merging** or **joining**.

While NumPy is excellent for numerical operations on single arrays, for structured data (like tables with columns and rows, similar to spreadsheets), the **Pandas** library is the go-to tool in Python. Pandas builds on NumPy and provides powerful data structures like `DataFrame`s, which are ideal for tabular data.



### Introduction to Pandas `DataFrame` (brief)

A Pandas `DataFrame` can be thought of as a spreadsheet or a SQL table. It's a 2-dimensional labeled data structure with columns of potentially different types. You can create DataFrames from dictionaries, lists of lists, or by reading data from files (CSV, Excel, etc.). We’ll be using pandas throughout the course, so you’ll have plenty of chances to learn more.



### The `pd.merge()` function

The `pd.merge()` function is the primary way to combine DataFrames based on common columns. It's conceptually similar to SQL JOIN operations.

**Syntax:** `pd.merge(left, right, on=None, how='inner')`

- `left`: The first DataFrame to merge.
- `right`: The second DataFrame to merge.
- `on`: (Optional) Column or list of columns to join on. If `None`, it defaults to the intersection of column names in `left` and `right`.
- `how`: (Optional) Type of merge to be performed. Common types include:
    - `'inner'` (default): Use intersection of keys from both frames.
    - `'left'` (left outer join): Use keys from left frame.
    - `'right'` (right outer join): Use keys from right frame.
    - `'outer'` (full outer join): Use union of keys from both frames.

**Tip (very important):** The column name(s) specified in `on` must exist with the same name in both DataFrames. If the key names differ, the merge will not succeed.

#### Inner merge

Let's say we have two DataFrames: one with student IDs and names, and another with student IDs and their scores. We want to combine them to get student names along with their scores.



```python
import pandas as pd

# DataFrame 1: Student Names
students_df = pd.DataFrame({
  "student_id": [101, 102, 103, 104],
  "name": ["Alice", "Bob", "Charlie", "David"]
})

# DataFrame 2: Student Scores
scores_df = pd.DataFrame({
  "student_id": [101, 103, 104, 105],
  "score": [85, 92, 78, 95]
})

print("Students DataFrame:")
print(students_df)
print("\nScores DataFrame:")
print(scores_df)

# Perform an inner merge
merged_df_inner = pd.merge(students_df, scores_df, on="student_id", how="inner")
print("\nInner Merged DataFrame (only common student_ids):")
print(merged_df_inner)
```

### Inspect, select, filter, and create columns

`.shape` reports rows and columns; selecting a list of column names keeps those columns. A Boolean condition selects rows. `.copy()` makes the filtered table independent before we add columns. Here a score is measured out of 100; dividing by 100 changes its unit to a proportion.


```python
print("Rows and columns:", readable_merge.shape)
print(readable_merge.dtypes)
selected = readable_merge[["name", "score"]]
high_scores = readable_merge.loc[readable_merge["score"] >= 80].copy()
high_scores["score_fraction"] = high_scores["score"] / 100
high_scores["points_above_80"] = high_scores["score"] - 80
high_scores
```

### Grouped means using the same table

Create a group from the existing scores, then calculate its mean. This demonstrates `groupby` syntax; a high-score group having a higher mean is true by construction and supplies no causal evidence. Missing scores must not silently become failures.


```python
comparison = pd.merge(students_df, scores_df, on="student_id", how="left")
comparison["scored_at_least_80"] = comparison["score"].ge(80).astype("boolean")
comparison.loc[comparison["score"].isna(), "scored_at_least_80"] = pd.NA
comparison.groupby("scored_at_least_80", dropna=False)["score"].agg(
    mean_score="mean", observed_scores="count"
)
```


#### Left, right, and outer merges

The `how` argument dictates which keys to include in the merged DataFrame:

-   **Left Merge (`how='left'`):** Includes all rows from the left DataFrame and matching rows from the right. If there's no match in the right DataFrame, `NaN` (Not a Number) is filled for the columns from the right DataFrame.
-   **Right Merge (`how='right'`):** Includes all rows from the right DataFrame and matching rows from the left. If there's no match in the left DataFrame, `NaN` is filled for the columns from the left DataFrame.
-   **Outer Merge (`how='outer'`):** Includes all rows from both DataFrames. Where there are no matching values, `NaN` is filled.

**Tip** The logic of merge can be tricky to remember at first. If you’re unsure, use `help(pd.merge)` to quickly check how it works.


**5. Key names differ in pd.merge**




```python
students_df = pd.DataFrame({
    "student_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"]
})

scores_df = pd.DataFrame({
    "id": [101, 103, 104],
    "score": [85, 92, 78]
})

# Bad KeyError (student_id not found in right DataFrame)
# merged = pd.merge(students_df, scores_df, on="student_id")

# Good #1: Rename the column

scores_df_r = scores_df.rename(columns={"id": "student_id"})
merged_1 = pd.merge(students_df, scores_df_r, on="student_id")
print(merged_1)

# Good #2: use left_on / right_on
merged_2 = pd.merge(students_df, scores_df, left_on="student_id", right_on="id")
print(merged_2)

```


**6. Using the wrong merge type**



```python
students_df = pd.DataFrame({
    "student_id": [101, 102],
    "name": ["Alice", "Bob"]
})
scores_df = pd.DataFrame({
    "student_id": [201, 202],
    "score": [85, 90]
})

# Bad: Inner merge gives empty DataFrame (no common IDs)
merged_inner = pd.merge(students_df, scores_df, on="student_id", how="inner")
print(merged_inner)   # Empty!

# Good: Use outer merge if you want to keep all data
merged_outer = pd.merge(students_df, scores_df, on="student_id", how="outer")
print(merged_outer)
```

Practice merging two DataFrames with `pd.merge`, and understand how how='left', 'right', and 'outer' change the result. Answer these questions (you do not need any coding for answering these):

```python
students_df = pd.DataFrame({
  "student_id": [101, 102, 103, 104],
  "name": ["Alice", "Bob", "Charlie", "David"]
})
scores_df = pd.DataFrame({
  "student_id": [101, 103, 104, 105],
  "score": [85, 92, 78, 95]
})

# Left merge
merged_df_left = pd.merge(students_df, scores_df, on="student_id", how="left")
print("\nLeft Merged DataFrame (all students from left, matching scores):")
print(merged_df_left)

# Right merge
merged_df_right = pd.merge(students_df, scores_df, on="student_id", how="right")
print("\nRight Merged DataFrame (all scores from right, matching students):")
print(merged_df_right)

# Outer merge
merged_df_outer = pd.merge(students_df, scores_df, on="student_id", how="outer")
print("\nOuter Merged DataFrame (all students and all scores):")
print(merged_df_outer)
```
